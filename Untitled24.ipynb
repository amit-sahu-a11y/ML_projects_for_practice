### Training a Simple Machine Learning Model

This example demonstrates a basic classification task using `scikit-learn`. We'll generate synthetic data, split it, train a Logistic Regression model, and evaluate its performance.

```python
# Import necessary libraries
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
from sklearn.datasets import make_classification
import pandas as pd
import numpy as np
```

#### 1. Generate Synthetic Data

We'll create a simple binary classification dataset with `make_classification`.

```python
# Generate synthetic dataset
X, y = make_classification(n_samples=1000, n_features=10, n_informative=5, n_redundant=0,
                           n_clusters_per_class=1, random_state=42)

# Convert to DataFrame for better readability
df = pd.DataFrame(X, columns=[f'feature_{i}' for i in range(X.shape[1])])
df['target'] = y

print("Dataset head:")
display(df.head())
print("\nTarget distribution:")
display(df['target'].value_counts())
```

#### 2. Split Data into Training and Testing Sets

It's crucial to split your data to evaluate the model on unseen examples.

```python
# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"Training set size: {X_train.shape[0]} samples")
print(f"Testing set size: {X_test.shape[0]} samples")
```

#### 3. Train a Simple Model (Logistic Regression)

We'll use a `LogisticRegression` model, a common choice for binary classification.

```python
# Initialize and train the Logistic Regression model
model = LogisticRegression(random_state=42, solver='liblinear') # 'liblinear' is good for small datasets
model.fit(X_train, y_train)

print("Model training complete.")
```

#### 4. Make Predictions on the Test Set

Now, let's see how our trained model performs on the data it hasn't seen before.

```python
# Make predictions on the test set
y_pred = model.predict(X_test)

print("First 10 actual labels:", y_test[:10])
print("First 10 predicted labels:", y_pred[:10])
```

#### 5. Evaluate the Model

We'll use accuracy score and a classification report to evaluate the model's performance.

```python
# Evaluate the model's performance
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)

print(f"Accuracy: {accuracy:.4f}")
print("\nClassification Report:")
print(report)
```

#### 6. Calculate Feature Importance

For a Logistic Regression model, the absolute values of the coefficients can indicate the importance of each feature. Larger absolute values mean a stronger impact on the prediction.

```python
# Get feature names from the DataFrame
feature_names = df.drop(columns=['target']).columns

# Get the coefficients from the trained model
# For binary classification, coef_ is usually a 2D array, take the first row.
coefficients = model.coef_[0]

# Create a DataFrame for better visualization
feature_importance = pd.DataFrame({'Feature': feature_names, 'Importance': coefficients})

# Sort by absolute importance to see the most influential features
feature_importance['Absolute Importance'] = np.abs(feature_importance['Importance'])
feature_importance = feature_importance.sort_values(by='Absolute Importance', ascending=False)

print("Feature Importance (sorted by absolute value of coefficient):")
display(feature_importance)
```

#### 7. Compare Model Performance After Dropping Least Important Features

Based on the feature importance, let's identify and drop the least important features (e.g., those with the smallest absolute coefficients) and retrain the model to see if performance changes.

```python
# Identify the features to drop (e.g., the 3 least important ones)
least_important_features = feature_importance.tail(3)['Feature'].tolist()
print(f"Dropping features: {least_important_features}")

# Create a new dataset without these features
X_reduced = df.drop(columns=least_important_features + ['target']).values
y_reduced = df['target'].values

# Split the reduced data into training and testing sets
X_train_reduced, X_test_reduced, y_train_reduced, y_test_reduced = train_test_split(
    X_reduced, y_reduced, test_size=0.3, random_state=42, stratify=y_reduced)

print(f"Reduced training set size: {X_train_reduced.shape[0]} samples")
print(f"Reduced testing set size: {X_test_reduced.shape[0]} samples")
```

#### 8. Retrain Model with Reduced Feature Set

```python
# Initialize and train a new Logistic Regression model with the reduced feature set
model_reduced = LogisticRegression(random_state=42, solver='liblinear')
model_reduced.fit(X_train_reduced, y_train_reduced)

print("Model training with reduced features complete.")
```

#### 9. Evaluate the New Model and Compare

```python
# Make predictions on the reduced test set
y_pred_reduced = model_reduced.predict(X_test_reduced)

# Evaluate the new model's performance
accuracy_reduced = accuracy_score(y_test_reduced, y_pred_reduced)
report_reduced = classification_report(y_test_reduced, y_pred_reduced)

print(f"Original Model Accuracy: {accuracy:.4f}")
print(f"Reduced Model Accuracy: {accuracy_reduced:.4f}")

print("\nOriginal Model Classification Report:")
print(report)

print("\nReduced Model Classification Report:")
print(report_reduced)

# Compare accuracies
if accuracy_reduced > accuracy:
    print("\nDropping least important features IMPROVED model accuracy!")
elif accuracy_reduced < accuracy:
    print("\nDropping least important features DECREASED model accuracy.")
else:
    print("\nDropping least important features had NO CHANGE on model accuracy.")
```
