## Understanding Transformers From Scratch

Transformers are a groundbreaking neural network architecture introduced in the 2017 paper "Attention Is All You Need" by Vaswani et al. They revolutionized sequence-to-sequence tasks, especially in Natural Language Processing (NLP), by replacing recurrent and convolutional layers with a mechanism called **self-attention**.

### Key Innovations of Transformers:

1.  **Self-Attention**: Allows the model to weigh the importance of different words in the input sequence when processing each word, capturing long-range dependencies more effectively than RNNs.
2.  **Parallelization**: Unlike RNNs, which process sequences step-by-step, Transformers can process all words in a sequence simultaneously, significantly speeding up training.
3.  **Positional Encoding**: Since self-attention alone doesn't capture word order, positional encodings are added to the input embeddings to inject information about the relative or absolute position of words in the sequence.

### 1. The Overall Architecture: Encoder-Decoder Structure

The Transformer model typically consists of an **Encoder** and a **Decoder**. Both are composed of identical layers stacked on top of each other.

#### The Encoder

The Encoder takes an input sequence (e.g., a sentence) and maps it to a sequence of continuous representations. It processes the input in parallel and aims to understand the context of each word.

A single Encoder layer is made up of two main sub-layers:

1.  **Multi-Head Self-Attention Mechanism**: This layer helps the encoder to weigh the importance of different words in the input sequence relative to each other.
2.  **Position-wise Feed-Forward Network**: A simple fully connected feed-forward network applied independently to each position.

#### The Decoder

The Decoder takes the output of the encoder and its own previous output to generate an output sequence (e.g., a translated sentence). It's designed to be auto-regressive, meaning it generates one token at a time, taking previously generated tokens as input.

A single Decoder layer has three main sub-layers:

1.  **Masked Multi-Head Self-Attention**: Similar to the encoder's self-attention, but it masks future positions to prevent the decoder from 'cheating' by looking at the tokens it's supposed to predict.
2.  **Multi-Head Encoder-Decoder Attention**: This layer performs attention over the output of the encoder stack, allowing the decoder to focus on relevant parts of the input sequence.
3.  **Position-wise Feed-Forward Network**: Same as in the encoder.

All sub-layers in both the encoder and decoder employ **residual connections** around them, followed by **layer normalization**.

### 2. The Core Mechanism: Self-Attention

Self-attention is the heart of the Transformer. It allows a word to consider all other words in the sequence to better understand its own context. For each word, it computes three vectors:

*   **Query (Q)**: Represents the current word being processed.
*   **Key (K)**: Represents all other words in the sequence that the current word might attend to.
*   **Value (V)**: Contains the actual content or information of all other words.

The attention mechanism then calculates a weighted sum of the Value vectors, where the weights are determined by the similarity between the Query and Key vectors. This is called **Scaled Dot-Product Attention**.

#### Scaled Dot-Product Attention Formula:

$$ \text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V $$

```python
import torch
import torch.nn as nn
import math

def scaled_dot_product_attention(query, key, value, mask=None):
    """Calculates scaled dot-product attention.
    Args:
        query: Query tensor (batch_size, num_heads, seq_len, d_k)
        key: Key tensor (batch_size, num_heads, seq_len, d_k)
        value: Value tensor (batch_size, num_heads, seq_len, d_v)
        mask: Optional mask tensor (batch_size, 1, 1, seq_len)
    Returns:
        output: Context vector (batch_size, num_heads, seq_len, d_v)
        attention_weights: Attention weights (batch_size, num_heads, seq_len, seq_len)
    """
    d_k = query.size(-1)
    scores = torch.matmul(query, key.transpose(-2, -1)) / math.sqrt(d_k)

    if mask is not None:
        # Apply the mask to fill positions with -inf where mask is 0
        scores = scores.masked_fill(mask == 0, -1e9)

    attention_weights = torch.softmax(scores, dim=-1)
    output = torch.matmul(attention_weights, value)

    return output, attention_weights

print("Scaled Dot-Product Attention function defined.")
```

### 3. Multi-Head Attention

Instead of performing a single attention function, Multi-Head Attention linearly projects the Queries, Keys, and Values `h` times with different, learned linear projections. For each of these projected versions, it then performs the attention function in parallel. The resulting `h` attention outputs are concatenated and once again linearly projected to produce the final output.

This allows the model to jointly attend to information from different representation subspaces at different positions. It's like having multiple independent experts looking at the same data from different angles.

```python
import torch.nn as nn

class MultiHeadAttention(nn.Module):
    def __init__(self, d_model, num_heads):
        super(MultiHeadAttention, self).__init__()
        assert d_model % num_heads == 0
        self.d_k = d_model // num_heads
        self.num_heads = num_heads
        self.w_q = nn.Linear(d_model, d_model)
        self.w_k = nn.Linear(d_model, d_model)
        self.w_v = nn.Linear(d_model, d_model)
        self.w_o = nn.Linear(d_model, d_model)

    def forward(self, query, key, value, mask=None):
        batch_size = query.size(0)

        # 1) Linear projections and reshape into (batch_size, num_heads, seq_len, d_k)
        query = self.w_q(query).view(batch_size, -1, self.num_heads, self.d_k).transpose(1, 2)
        key = self.w_k(key).view(batch_size, -1, self.num_heads, self.d_k).transpose(1, 2)
        value = self.w_v(value).view(batch_size, -1, self.num_heads, self.d_k).transpose(1, 2)

        # 2) Apply scaled dot-product attention
        x, attention_weights = scaled_dot_product_attention(query, key, value, mask)

        # 3) Concatenate heads and apply final linear layer
        x = x.transpose(1, 2).contiguous().view(batch_size, -1, self.num_heads * self.d_k)
        output = self.w_o(x)

        return output, attention_weights

print("MultiHeadAttention class defined.")
```

### 4. Positional Encoding

Since Transformers do not have recurrence or convolutions, they inherently lack an understanding of the order of words in a sequence. To address this, positional encodings are added to the input embeddings before they enter the encoder and decoder stacks. These encodings provide information about the relative or absolute position of each token in the sequence.

The original paper used sine and cosine functions of different frequencies for positional encoding:

$$ PE_{(pos, 2i)} = \sin(pos / 10000^{2i/d_{\text{model}}}) $$
$$ PE_{(pos, 2i+1)} = \cos(pos / 10000^{2i/d_{\text{model}}}) $$

```python
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super(PositionalEncoding, self).__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))

        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0) # Add batch dimension
        self.register_buffer('pe', pe)

    def forward(self, x):
        # Add positional encoding to input embeddings
        # x is (batch_size, seq_len, d_model)
        x = x + self.pe[:, :x.size(1), :]
        return x

print("PositionalEncoding class defined.")
```

### 5. Position-wise Feed-Forward Networks

Each encoder and decoder layer contains a simple, fully connected feed-forward network that is applied identically and independently to each position. This consists of two linear transformations with a ReLU activation in between.

$$ \text{FFN}(x) = \max(0, xW_1 + b_1)W_2 + b_2 $$

```python
class PositionwiseFeedForward(nn.Module):
    def __init__(self, d_model, d_ff, dropout=0.1):
        super(PositionwiseFeedForward, self).__init__()
        self.w_1 = nn.Linear(d_model, d_ff)
        self.w_2 = nn.Linear(d_ff, d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        return self.w_2(self.dropout(nn.functional.relu(self.w_1(x))))

print("PositionwiseFeedForward class defined.")
```

### 6. Layer Normalization and Residual Connections

Both the encoder and decoder layers use **residual connections** (also known as skip connections) around each of their sub-layers, followed by **layer normalization**. This helps stabilize training by allowing gradients to flow more easily through the network and mitigating the vanishing/exploding gradient problem.

```python
class LayerNorm(nn.Module):
    def __init__(self, features, eps=1e-6):
        super(LayerNorm, self).__init__()
        self.a_2 = nn.Parameter(torch.ones(features))
        self.b_2 = nn.Parameter(torch.zeros(features))
        self.eps = eps

    def forward(self, x):
        mean = x.mean(-1, keepdim=True)
        std = x.std(-1, keepdim=True)
        return self.a_2 * (x - mean) / (std + self.eps) + self.b_2

class SublayerConnection(nn.Module):
    """A residual connection followed by a layer norm."""
    def __init__(self, size, dropout):
        super(SublayerConnection, self).__init__()
        self.norm = LayerNorm(size)
        self.dropout = nn.Dropout(dropout)
        self.size = size

    def forward(self, x, sublayer):
        """Apply residual connection to any sublayer with the same size."""
        return x + self.dropout(sublayer(self.norm(x)))

print("LayerNorm and SublayerConnection classes defined.")
```

### 7. Putting It Together: The Encoder Layer

Now, let's combine these components to build a single Encoder layer.

```python
class EncoderLayer(nn.Module):
    """One layer of the encoder stack."""
    def __init__(self, d_model, num_heads, d_ff, dropout):
        super(EncoderLayer, self).__init__()
        self.self_attn = MultiHeadAttention(d_model, num_heads)
        self.feed_forward = PositionwiseFeedForward(d_model, d_ff, dropout)
        self.sublayer = nn.ModuleList([SublayerConnection(d_model, dropout) for _ in range(2)])
        self.d_model = d_model

    def forward(self, x, mask):
        # Apply multi-head self-attention with residual connection and layer norm
        x = self.sublayer[0](x, lambda x: self.self_attn(x, x, x, mask)[0])

        # Apply position-wise feed-forward network with residual connection and layer norm
        x = self.sublayer[1](x, self.feed_forward)
        return x

print("EncoderLayer class defined.")
```

This is a simplified, from-scratch breakdown of the core components of a Transformer. A full Transformer model would stack multiple `EncoderLayer` instances to form the `Encoder` and similarly build the `Decoder` with its additional masked attention and encoder-decoder attention layers.

### 8. The Encoder Stack

The full Encoder is a stack of `N` identical `EncoderLayer` instances. The input to the first encoder layer is the sum of the word embeddings and their corresponding positional encodings. The output of each layer is passed as input to the next layer.

```python
class Encoder(nn.Module):
    """Core encoder is a stack of N layers."""
    def __init__(self, layer, N):
        super(Encoder, self).__init__()
        self.layers = nn.ModuleList([layer for _ in range(N)])
        self.norm = LayerNorm(layer.d_model)

    def forward(self, x, mask):
        """Pass the input (and mask) through each layer in turn."""
        for layer in self.layers:
            x = layer(x, mask)
        return self.norm(x)

print("Encoder class defined.")
```

### 9. The Decoder Layer

The Decoder layer is more complex than the Encoder layer, as it includes an additional attention mechanism (Encoder-Decoder Attention) to attend over the output of the Encoder stack.

```python
class DecoderLayer(nn.Module):
    """One layer of the decoder stack."""
    def __init__(self, d_model, num_heads, d_ff, dropout):
        super(DecoderLayer, self).__init__()
        self.self_attn = MultiHeadAttention(d_model, num_heads)
        self.src_attn = MultiHeadAttention(d_model, num_heads) # Encoder-Decoder Attention
        self.feed_forward = PositionwiseFeedForward(d_model, d_ff, dropout)
        self.sublayer = nn.ModuleList([SublayerConnection(d_model, dropout) for _ in range(3)])

    def forward(self, x, memory, src_mask, tgt_mask):
        """x is the target sequence, memory is the encoder output."""
        m = memory

        # Masked Multi-Head Self-Attention for target sequence
        x = self.sublayer[0](x, lambda x: self.self_attn(x, x, x, tgt_mask)[0])

        # Multi-Head Encoder-Decoder Attention (Query from target, Key/Value from encoder output)
        x = self.sublayer[1](x, lambda x: self.src_attn(x, m, m, src_mask)[0])

        # Position-wise Feed-Forward Network
        x = self.sublayer[2](x, self.feed_forward)
        return x

print("DecoderLayer class defined.")
```

### 10. The Decoder Stack

Similar to the Encoder, the full Decoder is a stack of `N` identical `DecoderLayer` instances. It takes the output of the Encoder stack and the previously generated target sequence as input.

```python
class Decoder(nn.Module):
    """Generic N layer decoder with masking."""
    def __init__(self, layer, N):
        super(Decoder, self).__init__()
        self.layers = nn.ModuleList([layer for _ in range(N)])
        self.norm = LayerNorm(layer.d_model)

    def forward(self, x, memory, src_mask, tgt_mask):
        for layer in self.layers:
            x = layer(x, memory, src_mask, tgt_mask)
        return self.norm(x)

print("Decoder class defined.")
```

### 11. The Full Transformer Model

Finally, we can assemble all these components into the complete Transformer architecture. It consists of an Encoder, a Decoder, input/output embeddings, and positional encodings.

```python
class Transformer(nn.Module):
    """A standard Transformer model."""
    def __init__(self, src_vocab, tgt_vocab, N=6, d_model=512, d_ff=2048, num_heads=8, dropout=0.1):
        super(Transformer, self).__init__()
        # Embedding layers
        self.src_embed = nn.Sequential(nn.Embedding(src_vocab, d_model), PositionalEncoding(d_model, dropout))
        self.tgt_embed = nn.Sequential(nn.Embedding(tgt_vocab, d_model), PositionalEncoding(d_model, dropout))

        # Encoder and Decoder stacks
        self.encoder = Encoder(EncoderLayer(d_model, num_heads, d_ff, dropout), N)
        self.decoder = Decoder(DecoderLayer(d_model, num_heads, d_ff, dropout), N)

        # Final linear layer for output prediction
        self.generator = nn.Linear(d_model, tgt_vocab)

        # Initialize parameters with Glorot / fan_avg
        for p in self.parameters():
            if p.dim() > 1:
                nn.init.xavier_uniform_(p)

    def forward(self, src, tgt, src_mask, tgt_mask):
        """Take in and process masked src and target sequences."""
        encoder_output = self.encode(src, src_mask)
        decoder_output = self.decode(encoder_output, src_mask, tgt, tgt_mask)
        return self.generator(decoder_output)

    def encode(self, src, src_mask):
        return self.encoder(self.src_embed(src), src_mask)

    def decode(self, memory, src_mask, tgt, tgt_mask):
        return self.decoder(self.tgt_embed(tgt), memory, src_mask, tgt_mask)

print("Transformer class defined.")
```

## Minimal Training Loop for the Transformer Model

To demonstrate how the Transformer model can be used, let's create a minimal training loop. We'll use dummy data for simplicity, focusing on the mechanics of feeding data through the model, calculating loss, and performing optimization steps.

### 1. Utility Functions: Masks and Batching

Transformers require specific masking for both the source (encoder input) and target (decoder input) sequences.

*   **Source Mask (`src_mask`)**: Prevents the attention mechanism from attending to padding tokens in the input sequence.
*   **Target Mask (`tgt_mask`)**: Prevents the decoder from attending to future tokens in the target sequence (look-ahead mask) and also to padding tokens.

```python
def make_src_mask(src, pad_idx):
    """Create a mask to hide padding tokens in the source sequence."""
    # src shape: (batch_size, seq_len)
    # Returns: (batch_size, 1, 1, seq_len)
    src_mask = (src != pad_idx).unsqueeze(1).unsqueeze(2)
    return src_mask

def make_tgt_mask(tgt, pad_idx):
    """Create a combined mask for target sequence (padding and look-ahead)."""
    # tgt shape: (batch_size, seq_len)
    # Returns: (batch_size, 1, seq_len, seq_len)

    # Padding mask: Hide padding tokens
    tgt_pad_mask = (tgt != pad_idx).unsqueeze(1).unsqueeze(2) # (batch_size, 1, 1, seq_len)

    # Look-ahead mask: Hide future tokens (upper triangle)
    seq_len = tgt.size(1)
    tgt_lookahead_mask = torch.triu(torch.ones(seq_len, seq_len)).transpose(0, 1) # (seq_len, seq_len)
    tgt_lookahead_mask = tgt_lookahead_mask.bool().unsqueeze(0).unsqueeze(0) # (1, 1, seq_len, seq_len)

    # Combine masks: element-wise AND
    tgt_mask = tgt_pad_mask & tgt_lookahead_mask
    return tgt_mask

print("Mask utility functions defined.")
```

### 2. Dummy Data Generation

We'll create simple integer sequences to represent our source and target data. In a real scenario, these would come from tokenized text.

```python
import random

# Define vocabulary sizes and padding index
SRC_VOCAB_SIZE = 100
TGT_VOCAB_SIZE = 100
PAD_IDX = 0  # Assuming 0 is our padding token

# Model hyperparameters (can be adjusted)
N_LAYERS = 2
D_MODEL = 128
D_FF = 512
NUM_HEADS = 4
DROPOUT = 0.1

# Training hyperparameters
BATCH_SIZE = 2
MAX_SEQ_LEN = 10
NUM_EPOCHS = 5
LEARNING_RATE = 0.001

def generate_dummy_batch(batch_size, max_seq_len, vocab_size, pad_idx):
    src_len = random.randint(5, max_seq_len)
    tgt_len = random.randint(5, max_seq_len)

    # Generate random sequences, ensure they are not all padding
    src = torch.randint(1, vocab_size, (batch_size, src_len)) # avoid 0 for actual tokens
    tgt = torch.randint(1, vocab_size, (batch_size, tgt_len)) # avoid 0 for actual tokens

    # Pad sequences to max_seq_len, if necessary (though current generation ensures this somewhat)
    # For simplicity here, we'll keep lengths random up to max_seq_len
    # In a real scenario, you'd pad to a common length for the batch.

    return src, tgt

print("Dummy data generation function defined.")
```

### 3. Model Instantiation, Loss, and Optimizer

We instantiate our `Transformer` model and set up a standard `CrossEntropyLoss` for sequence prediction, along with the `Adam` optimizer.

```python
import torch.optim as optim

# Instantiate the Transformer model
model = Transformer(
    src_vocab=SRC_VOCAB_SIZE,
    tgt_vocab=TGT_VOCAB_SIZE,
    N=N_LAYERS,
    d_model=D_MODEL,
    d_ff=D_FF,
    num_heads=NUM_HEADS,
    dropout=DROPOUT
)

# Define loss function (ignoring padding index)
criterion = nn.CrossEntropyLoss(ignore_index=PAD_IDX)

# Define optimizer
optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)

print("Transformer model, criterion, and optimizer instantiated.")
```

### 4. The Training Loop

This loop simulates training over several epochs. In each epoch, we generate a dummy batch, create the necessary masks, perform a forward pass, calculate the loss, and update the model's weights.

```python
model.train() # Set the model to training mode

print(f"Starting training for {NUM_EPOCHS} epochs...")

for epoch in range(NUM_EPOCHS):
    # Generate dummy batch
    src, tgt = generate_dummy_batch(BATCH_SIZE, MAX_SEQ_LEN, SRC_VOCAB_SIZE, PAD_IDX)

    # Prepare target input and output
    # For training, tgt_input is the sequence shifted right (SOS token implicitly handled by embedding)
    # tgt_output is the original target sequence (what the model should predict)
    tgt_input = tgt[:, :-1] # Shift right, e.g., if tgt is [A, B, C, D], input is [A, B, C]
    tgt_output = tgt[:, 1:] # Target for prediction, e.g., output is [B, C, D]

    # Create masks
    src_mask = make_src_mask(src, PAD_IDX)
    tgt_mask = make_tgt_mask(tgt_input, PAD_IDX)

    # Zero gradients
    optimizer.zero_grad()

    # Forward pass
    # The forward method of Transformer expects src, tgt, src_mask, tgt_mask
    output = model(src, tgt_input, src_mask, tgt_mask)

    # Reshape output for CrossEntropyLoss: (batch_size * seq_len, vocab_size)
    # Reshape target: (batch_size * seq_len)
    loss = criterion(output.contiguous().view(-1, TGT_VOCAB_SIZE), tgt_output.contiguous().view(-1))

    # Backward pass and optimize
    loss.backward()
    optimizer.step()

    print(f"Epoch [{epoch+1}/{NUM_EPOCHS}], Loss: {loss.item():.4f}")

print("Training loop finished.")
```

### Summary of the Training Loop

This minimal training loop demonstrates the core steps:

1.  **Data Preparation**: Generating synthetic `src` and `tgt` sequences, and splitting `tgt` into `tgt_input` (for decoder's self-attention) and `tgt_output` (for loss calculation).
2.  **Mask Generation**: Creating `src_mask` to ignore padding in the encoder and `tgt_mask` for both padding and look-ahead in the decoder.
3.  **Forward Pass**: Feeding the data and masks through the `Transformer` model.
4.  **Loss Calculation**: Comparing the model's output with the ground truth `tgt_output` using `CrossEntropyLoss`.
5.  **Backward Pass and Optimization**: Calculating gradients and updating model parameters using `Adam`.

While this is a simplified example, it provides a functional overview of how to train a Transformer model for a sequence-to-sequence task. A real-world application would involve a more sophisticated data pipeline, possibly a learning rate scheduler, and evaluation metrics.
