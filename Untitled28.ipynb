```python
import pandas as pd
df=pd.read_csv('/content/sample_data/california_housing_test.csv')
df
```

```python
x=df.drop(columns='median_income',axis=0)
y=df['median_income']
```

```python
y
```

```python
class lr:
  def __init__(self):
    self.m=None
    self.b=None

  def fit(self,x,y):
    num=0
    den=0
    for i in range(x.shape[0]):
      num=num+(x[i]-x.mean()) * (y[i]-y.mean())
      den=den+(x[i]-x.mean())*(x[i]-x.mean())

    self.m=num/den
    self.b=y.mean()-(self.m*x.mean())
```

```python
l=lr()
l.fit(x,y)
```
