```python
import tensorflow
from keras import Sequential
from keras.layers import Dense
import pandas as pd
```

```python
df=pd.read_csv('/content/sample_data/california_housing_train.csv')
df
```

```python
x=df.drop(columns='median_income')
y=df['median_income']
```

```python
model=Sequential()
```

```python
model.add(Dense(3,activation='relu',input_dim=8))
model.add(Dense(1,activation='sigmoid'))
```

```python
# from keras.src.losses import mean_absolute_error
# from keras.src.ops import binary_crossentropy
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'],)
```

```python
model.fit(x,y,epochs=1000,validation_split=0.2)
```
