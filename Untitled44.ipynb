## Loops in Python

Loops are fundamental programming constructs that allow you to execute a block of code repeatedly. Python offers two primary types of loops: `for` loops and `while` loops, each suited for different scenarios.

### 1. The `for` Loop

#### What it does:
A `for` loop is used for iterating over a sequence (like a list, tuple, string, dictionary, or range) or other iterable objects. It executes a block of code once for each item in the sequence.

#### Syntax:
```python
for item in sequence:
    # code block to be executed
```

#### Example 1: Iterating over a list

````python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"I love {fruit}s!")
````

#### Example 2: Iterating using `range()`

The `range()` function is often used with `for` loops to generate a sequence of numbers.
- `range(stop)`: Generates numbers from 0 up to (but not including) `stop`.
- `range(start, stop)`: Generates numbers from `start` up to (but not including) `stop`.
- `range(start, stop, step)`: Generates numbers from `start` up to (but not including) `stop`, incrementing by `step`.

````python
# Loop from 0 to 4
print("Counting from 0 to 4:")
for i in range(5):
    print(i)

# Loop from 2 to 6
print("\nCounting from 2 to 6:")
for i in range(2, 7):
    print(i)

# Loop from 10 down to 0, stepping by 2
print("\nCounting down from 10 to 0 (by 2s):")
for i in range(10, -1, -2):
    print(i)
````

#### Example 3: Iterating over a string

````python
word = "Python"
print("\nCharacters in 'Python':")
for char in word:
    print(char)
````

### 2. The `while` Loop

#### What it does:
A `while` loop repeatedly executes a block of code as long as a given condition is true. It's best used when you don't know in advance how many times the loop needs to run, but rather when a certain condition needs to be met.

#### Syntax:
```python
while condition:
    # code block to be executed
    # make sure to modify the condition inside the loop
```

#### Example 1: Simple counter

````python
count = 0
print("Counting up to 3:")
while count < 4:
    print(count)
    count += 1 # Increment count to eventually make the condition false
````

#### Example 2: User input loop (until a specific input)

````python
print("\nGuess the magic word (hint: it's 'colab'):")
magic_word = "colab"
guess = ""

while guess != magic_word:
    guess = input("Enter your guess: ").lower()
    if guess != magic_word:
        print("Try again!")

print("Congratulations! You guessed the magic word!")
````

### Loop Control Statements

Python provides statements to alter the execution flow of loops:

- `break`: Terminates the loop entirely and transfers control to the statement immediately following the loop.
- `continue`: Skips the rest of the current iteration and moves to the next iteration of the loop.
- `pass`: A null operation; nothing happens when it executes. It's useful as a placeholder where a statement is syntactically required but you don't want any code to execute.

#### Example: `break` and `continue`

````python
print("\nUsing break:")
for i in range(10):
    if i == 5:
        print("Breaking loop at 5")
        break # Exit the loop when i is 5
    print(i)

print("\nUsing continue:")
for i in range(10):
    if i % 2 == 0: # If i is even
        continue # Skip the rest of the current iteration
    print(i) # This will only print odd numbers
````

These examples should give you a solid understanding of how to use `for` and `while` loops, along with loop control statements, in your Python code!
