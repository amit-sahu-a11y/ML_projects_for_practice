## Shallow Copy vs. Deep Copy in Python

In Python, when you copy an object, you have two main options: a **shallow copy** and a **deep copy**. The choice between them depends on whether the object contains other objects (e.g., lists within lists, dictionaries containing lists).

### Shallow Copy (`copy.copy()` or slicing for lists)

A shallow copy creates a new object, but it does *not* create copies of the nested objects within the original. Instead, it copies the references to these nested objects. This means that if you modify a nested object in the copy, the change will also be reflected in the original object (and vice-versa), because both the original and the copy are pointing to the *same* nested object.

It's like making a photocopy of a document that contains sticky notes. The photocopy is new, but the sticky notes on the photocopy are the *same physical sticky notes* that are on the original document. If you write on a sticky note on the photocopy, the writing also appears on the original document's sticky note.

### Deep Copy (`copy.deepcopy()`)

A deep copy, on the other hand, creates a new object and then recursively creates copies of *all* nested objects found in the original. This means that a deep copy is completely independent of the original object. Changes made to the deep copy (or its nested objects) will not affect the original object, and vice-versa.

Using the same analogy, a deep copy would be like making a photocopy of a document and then also making photocopies of all the sticky notes, placing the copied sticky notes on the copied document. Now, writing on a sticky note on the copied document will *not* affect the original document's sticky notes because they are entirely separate.

**When to use which?**
*   Use **shallow copy** when you have a flat object (no nested objects) or when you want the copy to reflect changes in the original's nested objects (or vice-versa).
*   Use **deep copy** when you have nested objects and you need a completely independent copy, ensuring that changes to the copy do not impact the original.

```python
import copy

print("--- Shallow Copy Example ---")

# Example 1: Flat List (Shallow copy works like deep copy here)
original_list_flat = [1, 2, 3]
shallow_copy_flat = original_list_flat.copy() # or list(original_list_flat) or original_list_flat[:]

print(f"Original (flat): {original_list_flat}")
print(f"Shallow copy (flat): {shallow_copy_flat}")

shallow_copy_flat.append(4)
print(f"Original after modifying shallow copy (flat): {original_list_flat}") # Original is unchanged
print(f"Shallow copy after modifying (flat): {shallow_copy_flat}")


# Example 2: Nested List (Demonstrates the behavior of shallow copy)
original_list_nested = [1, [2, 3], 4]
shallow_copy_nested = copy.copy(original_list_nested)

print(f"\nOriginal (nested): {original_list_nested}")
print(f"Shallow copy (nested): {shallow_copy_nested}")

# Modify an element in the top-level list of the shallow copy
shallow_copy_nested.append(5)
print(f"Original after appending to shallow copy (nested): {original_list_nested}") # Original unchanged
print(f"Shallow copy after appending (nested): {shallow_copy_nested}")

# Modify a NESTED element in the shallow copy
# This changes the original because both original and copy refer to the SAME nested list [2, 3]
shallow_copy_nested[1].append(99)
print(f"Original after modifying nested element in shallow copy (nested): {original_list_nested}") # Original IS changed!
print(f"Shallow copy after modifying nested element (nested): {shallow_copy_nested}")
```

```python
import copy

print("--- Deep Copy Example ---")

# Example 1: Flat List (Deep copy works like shallow copy here)
original_list_flat_deep = [10, 20, 30]
deep_copy_flat = copy.deepcopy(original_list_flat_deep)

print(f"Original (flat): {original_list_flat_deep}")
print(f"Deep copy (flat): {deep_copy_flat}")

deep_copy_flat.append(40)
print(f"Original after modifying deep copy (flat): {original_list_flat_deep}") # Original is unchanged
print(f"Deep copy after modifying (flat): {deep_copy_flat}")


# Example 2: Nested List (Demonstrates the behavior of deep copy)
original_list_nested_deep = [10, [20, 30], 40]
deep_copy_nested = copy.deepcopy(original_list_nested_deep)

print(f"\nOriginal (nested): {original_list_nested_deep}")
print(f"Deep copy (nested): {deep_copy_nested}")

# Modify an element in the top-level list of the deep copy
deep_copy_nested.append(50)
print(f"Original after appending to deep copy (nested): {original_list_nested_deep}") # Original unchanged
print(f"Deep copy after appending (nested): {deep_copy_nested}")

# Modify a NESTED element in the deep copy
# This does NOT change the original because the nested list [20, 30] was also copied
deep_copy_nested[1].append(999)
print(f"Original after modifying nested element in deep copy (nested): {original_list_nested_deep}") # Original IS NOT changed!
print(f"Deep copy after modifying nested element (nested): {deep_copy_nested}")
```
