### Simple Machine Learning Model Example: Linear Regression

Let's train a basic linear regression model using `scikit-learn`. We'll generate some synthetic data, fit a model to it, and then make predictions.

```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
# Generate synthetic data
X = np.random.rand(100, 1) * 10  # 100 samples, 1 feature
y = 2 * X + 1 + np.random.randn(100, 1) * 2 # y = 2x + 1 + noise

print("Generated X (first 5 values):\n", X[:5])
print("Generated y (first 5 values):\n", y[:5])
```

```python
# Create and train the Linear Regression model
model = LinearRegression()
model.fit(X, y)

print(f"Model coefficients: {model.coef_[0][0]:.2f}")
print(f"Model intercept: {model.intercept_[0]:.2f}")
```

```python
# Make predictions
X_new = np.array([[0], [2], [5], [8], [10]])
y_pred = model.predict(X_new)

print("\nNew X values for prediction:\n", X_new)
print("Predicted y values:\n", y_pred)
```

```python
# Visualize the results
fig = plt.figure(figsize=(10, 6))
plt.scatter(X, y, label='Original Data')
plt.plot(X, model.predict(X), color='red', label='Regression Line')
plt.scatter(X_new, y_pred, color='green', marker='x', s=100, label='New Predictions')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Linear Regression Model')
plt.legend()
plt.grid(True)
plt.show()
```
