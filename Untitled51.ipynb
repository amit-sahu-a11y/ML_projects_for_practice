## Linear Regression from Scratch: A Deep Dive

Let's break down the Linear Regression algorithm from its fundamental principles to how it practically finds the best-fit line, including a basic implementation.

### 1. What is Linear Regression?

Linear Regression is a statistical method used for modeling the relationship between a scalar dependent variable (target) and one or more explanatory variables (features). It assumes a linear relationship, meaning the change in the dependent variable is proportional to the change in the independent variable(s).

**Goal:** To find the line (or hyperplane in higher dimensions) that best fits the data, minimizing the distance between the line and all data points. This line is then used to predict the target variable for new, unseen data.

### 2. The Equation of the Line

For **Simple Linear Regression** (one feature):

`y_pred = mx + c`

Where:
*   `y_pred` is the predicted value of the dependent variable.
*   `x` is the independent variable (feature).
*   `m` is the **slope** (or coefficient), representing how much `y_pred` changes for a unit change in `x`.
*   `c` is the **y-intercept**, representing the value of `y_pred` when `x` is 0.

For **Multiple Linear Regression** (multiple features):

`y_pred = m1*x1 + m2*x2 + ... + mn*xn + c`

Or, in vectorized form:

`y_pred = X * W + c`

Where:
*   `X` is the matrix of features.
*   `W` is the vector of weights (coefficients).
*   `c` is the bias (intercept).

### 3. How to Find the 'Best-Fit' Line: The Cost Function

How do we determine which line is 'best'? We need a way to measure how well a given line fits our data. This is done using a **Cost Function** (or Loss Function).

The most common cost function for linear regression is the **Mean Squared Error (MSE)**.

**Intuition:** For each data point, we calculate the difference between its actual `y` value and the `y_pred` value predicted by our line. We then square this difference to ensure positive values and to penalize larger errors more heavily. Finally, we take the average of these squared differences across all data points.

**Formula:**

`MSE = (1/N) * Σ(y_actual_i - y_pred_i)²`

Where:
*   `N` is the total number of data points.
*   `y_actual_i` is the actual value for the i-th data point.
*   `y_pred_i` is the predicted value for the i-th data point (`m*x_i + c`).
*   `Σ` denotes summation over all data points.

**Objective:** The goal of the linear regression algorithm is to find the values of `m` and `c` that **minimize** the MSE. A lower MSE indicates a better-fitting line.

### 4. Optimization: Finding the Minimum MSE (Gradient Descent)

Minimizing the MSE means finding the 'bottom of the valley' on a cost landscape (a plot of MSE versus different `m` and `c` values). An iterative optimization algorithm called **Gradient Descent** is typically used.

**How it works (Simplified):**
1.  **Start with random `m` and `c` values.**
2.  **Calculate the gradient:** Determine the slope of the cost function with respect to `m` and `c`. The gradient tells us the direction of the steepest ascent on the cost landscape.
3.  **Update `m` and `c`:** Move `m` and `c` in the *opposite* direction of the gradient (downhill) by a small step size, controlled by a **learning rate**. This step size determines how aggressively we approach the minimum.
    *   `m_new = m_old - learning_rate * (∂MSE/∂m)`
    *   `c_new = c_old - learning_rate * (∂MSE/∂c)`
4.  **Repeat steps 2 and 3** for a fixed number of iterations or until the change in `m` and `c` (or the MSE itself) becomes very small (convergence).

**The Derivatives (Gradients):**

*   `∂MSE/∂m = (-2/N) * Σ(x_i * (y_actual_i - y_pred_i))`
*   `∂MSE/∂c = (-2/N) * Σ(y_actual_i - y_pred_i))`

### 5. Assumptions of Linear Regression

Linear regression makes several assumptions about the data:
*   **Linearity:** The relationship between features and target is linear.
*   **Independence:** Observations are independent of each other.
*   **Homoscedasticity:** The variance of the residuals (errors) is constant across all levels of the independent variables.
*   **Normality:** The residuals are normally distributed.
*   **No Multicollinearity:** Independent variables are not highly correlated with each other (for multiple linear regression).

### 6. Implementing Linear Regression from Scratch (using our Pizza Example)

Let's implement a simple linear regression model using gradient descent for our pizza diameter vs. price example.

```python
import numpy as np
import matplotlib.pyplot as plt

# Our pizza data
X_scratch = np.array([8, 10, 12, 14]) # Diameter
y_scratch = np.array([10, 12, 14, 16]) # Price

# Hyperparameters
learning_rate = 0.001 # Reduced learning rate
iterations = 10000 # Increased iterations for convergence

# Initial parameters (randomly or set to 0)
m = 0.0 # slope
c = 0.0 # intercept

N = len(X_scratch) # Number of data points

# Store costs to visualize convergence
cost_history = []

print("Starting Gradient Descent...")

# Gradient Descent Loop
for i in range(iterations):
    # Calculate predictions
    y_pred = m * X_scratch + c

    # Calculate the cost (MSE)
    cost = (1/N) * np.sum((y_scratch - y_pred)**2)
    cost_history.append(cost)

    # Calculate gradients
    dm = (-2/N) * np.sum(X_scratch * (y_scratch - y_pred))
    dc = (-2/N) * np.sum(y_scratch - y_pred)

    # Update parameters
    m = m - learning_rate * dm
    c = c - learning_rate * dc

    if i % (iterations / 10) == 0: # Print progress 10 times
        print(f"Iteration {i}: Cost = {cost:.4f}, m = {m:.4f}, c = {c:.4f}")

print("\nGradient Descent Finished.")
print(f"Optimal m: {m:.4f}")
print(f"Optimal c: {c:.4f}")

# Plot the cost history
plt.figure(figsize=(10, 5))
plt.plot(range(iterations), cost_history)
plt.title('Cost Function (MSE) Over Iterations')
plt.xlabel('Iterations')
plt.ylabel('MSE Cost')
plt.grid(True)
plt.show()

# Plot the final regression line
plt.figure(figsize=(8, 6))
plt.scatter(X_scratch, y_scratch, color='blue', label='Actual Prices')
plt.plot(X_scratch, m * X_scratch + c, color='red', label='Regression Line (From Scratch)')
plt.title('Pizza Diameter vs. Price (Linear Regression From Scratch)')
plt.xlabel('Diameter (inches)')
plt.ylabel('Price ($)')
plt.grid(True)
plt.legend()
plt.show()

# Example prediction
new_pizza_diameter = 18
predicted_price_scratch = m * new_pizza_diameter + c
print(f"\nPredicted price for a {new_pizza_diameter}-inch pizza (from scratch model): ${predicted_price_scratch:.2f}")
```

### 7. Generating a Larger Synthetic Dataset

To better demonstrate the capabilities of linear regression and for comparison with libraries like Scikit-learn, let's create a larger, synthetic dataset with a clear linear relationship and some added noise.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42) # for reproducibility

# Generate a larger synthetic dataset
X_synth = 2 * np.random.rand(100, 1) # 100 data points, 1 feature
y_synth = 4 + 3 * X_synth + np.random.randn(100, 1) # y = 4 + 3X + noise

# Visualize the synthetic data
plt.figure(figsize=(10, 6))
plt.scatter(X_synth, y_synth, c='blue', alpha=0.7)
plt.title('Synthetic Dataset for Linear Regression')
plt.xlabel('Feature (X)')
plt.ylabel('Target (y)')
plt.grid(True)
plt.show()

print(f"Shape of X_synth: {X_synth.shape}")
print(f"Shape of y_synth: {y_synth.shape}")
print("First 5 rows of X_synth:\n", X_synth[:5])
print("First 5 rows of y_synth:\n", y_synth[:5])
```

### 8. Linear Regression with Scikit-learn (on Synthetic Data)

Now, let's use a widely-adopted machine learning library, Scikit-learn, to perform linear regression on our synthetic dataset. This will provide a professional implementation for comparison and demonstrate its ease of use.

```python
# Import LinearRegression from Scikit-learn
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Create a Linear Regression model instance
model_sklearn = LinearRegression()

# Train the model on our synthetic data
model_sklearn.fit(X_synth, y_synth)

# Make predictions using the trained model
y_pred_sklearn = model_sklearn.predict(X_synth)

# Evaluate the model (e.g., using Mean Squared Error)
mse_sklearn = mean_squared_error(y_synth, y_pred_sklearn)

print(f"Scikit-learn Model - Intercept: {model_sklearn.intercept_[0]:.4f}")
print(f"Scikit-learn Model - Coefficient: {model_sklearn.coef_[0][0]:.4f}")
print(f"Scikit-learn Model - Mean Squared Error (MSE): {mse_sklearn:.4f}")
```

```python
# Visualize the synthetic data and the Scikit-learn regression line
plt.figure(figsize=(10, 6))
plt.scatter(X_synth, y_synth, c='blue', alpha=0.7, label='Actual Data')
plt.plot(X_synth, y_pred_sklearn, color='red', linewidth=2, label='Scikit-learn Regression Line')
plt.title('Synthetic Data with Scikit-learn Linear Regression')
plt.xlabel('Feature (X)')
plt.ylabel('Target (y)')
plt.grid(True)
plt.legend()
plt.show()
```

### 9. Comparing Scratch Implementation with Scikit-learn

Now, let's bring both our from-scratch linear regression model and the Scikit-learn model together to compare their performance and insights on the synthetic dataset.

```python
# First, let's get predictions from our scratch model for the synthetic data
# We'll use the optimal 'm' and 'c' derived from the original scratch model on the pizza data.
# For a proper comparison on the *synthetic* data, we should re-train the scratch model on X_synth and y_synth.
# Let's re-run the scratch gradient descent on the synthetic data.

# Hyperparameters (can be tuned)
learning_rate_synth = 0.01 # Adjusted learning rate for synthetic data
iterations_synth = 100000  # More iterations for better convergence

# Initial parameters for scratch model on synthetic data
m_scratch_synth = 0.0
c_scratch_synth = 0.0

N_synth = len(X_synth)

for i in range(iterations_synth):
    y_pred_scratch_synth = m_scratch_synth * X_synth + c_scratch_synth

    # Gradients
    dm_scratch_synth = (-2/N_synth) * np.sum(X_synth * (y_synth - y_pred_scratch_synth))
    dc_scratch_synth = (-2/N_synth) * np.sum(y_synth - y_pred_scratch_synth)

    # Update parameters
    m_scratch_synth = m_scratch_synth - learning_rate_synth * dm_scratch_synth
    c_scratch_synth = c_scratch_synth - learning_rate_synth * dc_scratch_synth

# Make predictions with the re-trained scratch model
y_pred_scratch_synth = m_scratch_synth * X_synth + c_scratch_synth

# Calculate MSE for the scratch model on synthetic data
mse_scratch_synth = mean_squared_error(y_synth, y_pred_scratch_synth)

print(f"Scratch Model (on Synthetic Data) - Intercept: {c_scratch_synth:.4f}")
print(f"Scratch Model (on Synthetic Data) - Coefficient: {m_scratch_synth:.4f}")
print(f"Scratch Model (on Synthetic Data) - Mean Squared Error (MSE): {mse_scratch_synth:.4f}")

print("\nScikit-learn Model - Intercept: {model_sklearn.intercept_[0]:.4f}")
print(f"Scikit-learn Model - Coefficient: {model_sklearn.coef_[0][0]:.4f}")
print(f"Scikit-learn Model - Mean Squared Error (MSE): {mse_sklearn:.4f}")
```

```python
# Visualize both regression lines
plt.figure(figsize=(12, 8))
plt.scatter(X_synth, y_synth, c='blue', alpha=0.7, label='Actual Data')
plt.plot(X_synth, y_pred_sklearn, color='red', linewidth=2, label='Scikit-learn Regression Line')
plt.plot(X_synth, y_pred_scratch_synth, color='green', linestyle='--', linewidth=2, label='Scratch Regression Line')

plt.title('Comparison of Scratch vs. Scikit-learn Linear Regression on Synthetic Data')
plt.xlabel('Feature (X)')
plt.ylabel('Target (y)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
from sklearn.metrics import r2_score

# Calculate R-squared for the scratch model
r2_scratch_synth = r2_score(y_synth, y_pred_scratch_synth)

# Calculate R-squared for the Scikit-learn model
r2_sklearn = r2_score(y_synth, y_pred_sklearn)

print(f"Scratch Model (on Synthetic Data) - R-squared: {r2_scratch_synth:.4f}")
print(f"Scikit-learn Model - R-squared: {r2_sklearn:.4f}")
```
