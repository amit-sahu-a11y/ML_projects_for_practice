### Simple Generative Adversarial Network (GAN) Example

This code demonstrates a basic GAN architecture. A GAN consists of two neural networks:

1.  **Generator:** Learns to create new data instances that resemble the training data.
2.  **Discriminator:** Learns to distinguish between real data from the training set and fake data created by the generator.

The two networks are trained simultaneously in a zero-sum game: the generator tries to fool the discriminator, and the discriminator tries to not be fooled.

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
import matplotlib.pyplot as plt

print(f"TensorFlow Version: {tf.__version__}")
```

#### 1. Prepare Data (Simple 1D data for demonstration)

```python
# Generate some simple 'real' data (e.g., samples from a normal distribution)
def generate_real_samples(n_samples):
    X = np.random.randn(n_samples) * 2 + 5 # Mean 5, Std Dev 2
    y = np.ones((n_samples, 1)) # Label as 'real'
    return X, y

# Visualize real data
real_data, _ = generate_real_samples(1000)
plt.hist(real_data, bins=50)
plt.title('Distribution of Real Data')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

#### 2. Define the Discriminator Model

The discriminator is a binary classifier that takes an input (real or fake data) and outputs a probability of it being real.

```python
def define_discriminator(n_inputs=1):
    input_layer = layers.Input(shape=(n_inputs,))
    dense1 = layers.Dense(25, activation='relu', kernel_initializer='he_uniform')(input_layer)
    output_layer = layers.Dense(1, activation='sigmoid')(dense1)
    model = models.Model(inputs=input_layer, outputs=output_layer)
    # Compile model
    model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
    return model

# Create and summarize the discriminator
d_model = define_discriminator()
d_model.summary()
```

#### 3. Define the Generator Model

The generator takes a random noise vector as input and produces data that resembles the real data distribution.

```python
def define_generator(latent_dim, n_outputs=1):
    input_layer = layers.Input(shape=(latent_dim,))
    dense1 = layers.Dense(15, activation='relu', kernel_initializer='he_uniform')(input_layer)
    output_layer = layers.Dense(n_outputs, activation='linear')(dense1)
    model = models.Model(inputs=input_layer, outputs=output_layer)
    return model

# Create and summarize the generator
latent_dim = 5 # Dimension of the random noise input
g_model = define_generator(latent_dim)
g_model.summary()
```

#### 4. Define the GAN Model

The GAN model combines the generator and discriminator. When training the GAN, we fix the discriminator's weights and only train the generator to produce outputs that the discriminator classifies as 'real'.

```python
def define_gan(generator, discriminator):
    # Make weights in the discriminator not trainable when training the GAN
    discriminator.trainable = False

    # Connect them: generator output is discriminator input
    gan_output = discriminator(generator.output)
    model = models.Model(inputs=generator.input, outputs=gan_output)

    # Compile model
    model.compile(loss='binary_crossentropy', optimizer='adam')
    return model

# Create and summarize the GAN
gan_model = define_gan(g_model, d_model)
gan_model.summary()
```

#### 5. Training the GAN

The training involves alternating updates for the discriminator and the generator.

```python
# Generate random points in latent space as input for the generator
def generate_latent_points(latent_dim, n_samples):
    x_input = np.random.randn(latent_dim * n_samples)
    x_input = x_input.reshape(n_samples, latent_dim)
    return x_input

# Use the generator to generate 'fake' samples
def generate_fake_samples(generator, latent_dim, n_samples):
    x_input = generate_latent_points(latent_dim, n_samples)
    X = generator.predict(x_input, verbose=0)
    y = np.zeros((n_samples, 1)) # Label as 'fake'
    return X, y

# Train the generator and discriminator
def train(g_model, d_model, gan_model, latent_dim, n_epochs=5000, n_batch=128):
    bat_per_epo = int(np.ceil(1000 / n_batch)) # Assuming 1000 real samples
    half_batch = int(n_batch / 2)

    # Manually enumerate epochs
    for i in range(n_epochs):
        # Get randomly selected 'real' samples
        X_real, y_real = generate_real_samples(half_batch)
        # Generate 'fake' samples
        X_fake, y_fake = generate_fake_samples(g_model, latent_dim, half_batch)

        # Reshape X_real and X_fake for discriminator training and evaluation
        X_real_reshaped = X_real.reshape(half_batch, 1)
        X_fake_reshaped = X_fake.reshape(half_batch, 1)

        # Convert NumPy arrays to TensorFlow tensors
        X_real_tf = tf.convert_to_tensor(X_real_reshaped, dtype=tf.float32)
        y_real_tf = tf.convert_to_tensor(y_real, dtype=tf.float32)
        X_fake_tf = tf.convert_to_tensor(X_fake_reshaped, dtype=tf.float32)
        y_fake_tf = tf.convert_to_tensor(y_fake, dtype=tf.float32)

        # Train the discriminator on real and fake samples
        d_loss_real, _ = d_model.train_on_batch(X_real_tf, y_real_tf)
        d_loss_fake, _ = d_model.train_on_batch(X_fake_tf, y_fake_tf)
        d_loss = 0.5 * (d_loss_real + d_loss_fake)

        # Prepare points in latent space as input for the generator
        X_gan = generate_latent_points(latent_dim, n_batch)
        # Create inverted labels for the fake samples (i.e., 'real')
        y_gan = np.ones((n_batch, 1))

        # Convert NumPy arrays to TensorFlow tensors for GAN training
        X_gan_tf = tf.convert_to_tensor(X_gan, dtype=tf.float32)
        y_gan_tf = tf.convert_to_tensor(y_gan, dtype=tf.float32)

        # Update the generator via the discriminator's error
        g_loss = gan_model.train_on_batch(X_gan_tf, y_gan_tf)

        # Summarize model performance
        if (i+1) % (n_epochs / 10) == 0:
            print(f'>Epoch {i+1}, d_loss=[{d_loss:.3f}], g_loss=[{g_loss:.3f}]')

            # Evaluate the discriminator on real and fake data at intervals
            _, acc_real = d_model.evaluate(X_real_tf, y_real_tf, verbose=0)
            _, acc_fake = d_model.evaluate(X_fake_tf, y_fake_tf, verbose=0)
            print(f'Accuracy Real: {acc_real*100:.2f}%, Fake: {acc_fake*100:.2f}%')

# Train the GAN model
print("\nStarting GAN training...")
train(g_model, d_model, gan_model, latent_dim)
```

#### 6. Evaluate and Visualize Generated Samples

After training, we can use the generator to produce new samples and compare their distribution to the real data.

```python
# Generate samples after training
latent_points = generate_latent_points(latent_dim, 1000)
generated_samples = g_model.predict(latent_points)

# Plot the generated samples and compare with real data
plt.figure(figsize=(10, 6))
plt.hist(real_data, bins=50, alpha=0.5, label='Real Data', color='blue')
plt.hist(generated_samples, bins=50, alpha=0.5, label='Generated Data', color='red')
plt.title('Distribution of Real vs. Generated Data')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

### Addressing GAN Imbalance: Adjusting Learning Rates

As observed, the discriminator quickly learns to distinguish between real and fake samples, making it too powerful. To give the generator a better chance to learn, we can adjust the learning rates for each model. A common strategy is to:

1.  Use a **lower learning rate for the discriminator** (e.g., `1e-4`).
2.  Use a **slightly higher learning rate for the generator** (e.g., `2e-4` or keep default `1e-3` if discriminator is very strong).

Let's re-define the optimizers and recompile the models.

```python
# Define new optimizers with adjusted learning rates
d_optimizer = tf.keras.optimizers.Adam(learning_rate=0.0002)
g_optimizer = tf.keras.optimizers.Adam(learning_rate=0.0002) # Often set to be equal or slightly higher than d's when d is strong

# Recompile the discriminator model with the new optimizer
d_model.compile(loss='binary_crossentropy', optimizer=d_optimizer, metrics=['accuracy'])

# Recompile the GAN model with the new optimizer (this affects the generator's training)
gan_model.compile(loss='binary_crossentropy', optimizer=g_optimizer)

print("Discriminator and GAN models recompiled with adjusted learning rates.")
```
