## Displaying Outliers

Outliers are data points that significantly differ from other observations. They can occur due to measurement errors, data entry errors, or genuinely rare events. Identifying and understanding outliers is crucial in data analysis as they can skew results and mislead conclusions.

One common method to detect outliers is the **Interquartile Range (IQR)** method.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create a sample DataFrame with some potential outliers
data = {
    'Value': [10, 12, 11, 13, 10, 14, 15, 12, 100, 11, 9, 13, 14, 10, -5, 12],
    'Category': ['A', 'A', 'A', 'A', 'B', 'B', 'B', 'B', 'A', 'A', 'B', 'B', 'A', 'A', 'B', 'A']
}
df = pd.DataFrame(data)

print("Sample DataFrame:")
display(df)
```

### Outlier Detection using the IQR Method

The IQR method defines outliers as any data points that fall below `Q1 - 1.5 * IQR` or above `Q3 + 1.5 * IQR`.

-   **Q1**: The first quartile (25th percentile).
-   **Q3**: The third quartile (75th percentile).
-   **IQR**: Interquartile Range (Q3 - Q1).

```python
# Calculate Q1, Q3, and IQR for the 'Value' column
Q1 = df['Value'].quantile(0.25)
Q3 = df['Value'].quantile(0.75)
IQR = Q3 - Q1

# Define outlier bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1: {Q1}")
print(f"Q3: {Q3}")
print(f"IQR: {IQR}")
print(f"Lower Bound (Outlier threshold): {lower_bound}")
print(f"Upper Bound (Outlier threshold): {upper_bound}")

# Identify outliers
outliers = df[(df['Value'] < lower_bound) | (df['Value'] > upper_bound)]

print("\nIdentified Outliers:")
display(outliers)
```

### Visualizing Outliers

Box plots are excellent for visualizing the distribution of numerical data and clearly showing outliers. Data points outside the 'whiskers' of the box plot are typically considered outliers.

```python
# Box plot to visualize outliers
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['Value'], color='skyblue')
plt.title('Box Plot of Values to Show Outliers')
plt.ylabel('Value')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

A scatter plot can also be useful, especially when comparing two variables or when you want to highlight the specific outlier points in relation to the rest of the data.

```python
# Scatter plot to highlight outliers
plt.figure(figsize=(10, 6))
sns.scatterplot(x=df.index, y=df['Value'], color='blue', label='Normal Data')
sns.scatterplot(x=outliers.index, y=outliers['Value'], color='red', marker='X', s=100, label='Outliers')

# Optional: Add a line for the upper and lower bounds
plt.axhline(y=upper_bound, color='green', linestyle='--', label='Upper Bound')
plt.axhline(y=lower_bound, color='orange', linestyle='--', label='Lower Bound')

plt.title('Scatter Plot with Highlighted Outliers')
plt.xlabel('Index')
plt.ylabel('Value')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

These visualizations help in understanding the extent and nature of outliers in your dataset.
