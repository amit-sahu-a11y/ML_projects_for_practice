```python
class cllg:
  def __init__(self,stu_name,math_marks,english_marks,science_marks):
    self.math=math_marks
    self.science=science_marks
    self.english=english_marks
    name=stu_name
    print(name)
    print("ITM UNIVERSITY")

  def avg(self):
    avg=((self.math+self.science+self.english)/3)
    print(avg)
```

```python
class account:
  def __init__(self,balance,account_name):
    print('welcome to bank...')
    self.balance=balance
    self.__acc_name=account_name

  def credit(self,amount):
    # self.credit_amount=credit_amount
    self.balance += amount
    print(amount,"are credit in your account ",self.__acc_name)

  def debit(self,amount):
    self.balance -=amount
    print(amount,"are debit in your account ",self.__acc_name)

  def get_balance(self):
    print('Your bank balance is ',self.balance,'on this account number ',self.__acc_name)


sol=account(10000,124)
sol.credit(5000)
sol.get_balance()
sol.debit(10000)
sol.get_balance()
# del sol
# print(sol.__acc_name)
sol.get_balance()
```
