```python
my_list = [10, 25, 5, 30, 15, 20]
d = 18

filtered_list = [x for x in my_list if x > d]

print(f"Original list: {my_list}")
print(f"Value of d: {d}")
print(f"Values greater than d: {filtered_list}")
```

```python
sorted_my_list_desc = sorted(my_list, reverse=True)
print(f"Original list sorted in descending order: {sorted_my_list_desc}")
```

```python
sorted_filtered_list = sorted(filtered_list)

print(f"Sorted filtered list (ascending): {sorted_filtered_list}")
```

```python
average_value = sum(sorted_filtered_list) / len(sorted_filtered_list)

print(f"The average of the sorted filtered list is: {average_value}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create a figure with two subplots
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Plotting the original list
axes[0].bar(np.arange(len(my_list)), my_list, color='skyblue')
axes[0].set_title('Original List (my_list)')
axes[0].set_xlabel('Index')
axes[0].set_ylabel('Value')
axes[0].set_xticks(np.arange(len(my_list)))
axes[0].set_xticklabels(my_list) # Label bars with their values

# Plotting the filtered list
axes[1].bar(np.arange(len(filtered_list)), filtered_list, color='lightcoral')
axes[1].set_title('Filtered List (values > d)')
axes[1].set_xlabel('Index')
axes[1].set_ylabel('Value')
axes[1].set_xticks(np.arange(len(filtered_list)))
axes[1].set_xticklabels(filtered_list) # Label bars with their values

plt.tight_layout() # Adjust layout to prevent overlapping titles/labels
plt.show()
```
