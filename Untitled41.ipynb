## Understanding Transformers

Transformers are a type of neural network architecture that have revolutionized the field of Natural Language Processing (NLP). They are particularly good at handling sequential data, like text, by using a mechanism called 'attention' to weigh the importance of different parts of the input sequence. This allows them to capture long-range dependencies in the data much more effectively than previous architectures like LSTMs or GRUs.

Key applications of transformers in data science include:
*   **Text Classification**: Sentiment analysis, spam detection, topic labeling.
*   **Named Entity Recognition (NER)**: Identifying and classifying entities like names, organizations, locations.
*   **Machine Translation**: Translating text from one language to another.
*   **Question Answering**: Answering questions based on a given text.
*   **Text Generation**: Creating human-like text, such as summaries, articles, or even creative writing.
*   **Summarization**: Condensing long texts into shorter, coherent summaries.

The most widely used library for working with transformers is [Hugging Face Transformers](https://huggingface.co/docs/transformers/index), which provides pre-trained models and tools to fine-tune them for various tasks.

```python
# First, install the Hugging Face Transformers library if you haven't already.
!pip install --upgrade transformers tensorflow

import tensorflow as tf
from transformers import AutoTokenizer, TFAutoModelForSequenceClassification

print("Libraries imported successfully!")
```

### Text Classification Example

Let's use a pre-trained `DistilBERT` model from Hugging Face for a simple text classification task. We'll classify sentiment (positive/negative) for a few example sentences.

```python
# Load a pre-trained tokenizer and model
model_name = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = TFAutoModelForSequenceClassification.from_pretrained(model_name)

# Define some sample sentences
texts = [
    "This movie is absolutely fantastic!",
    "I really disliked the ending, it was so predictable.",
    "The food was okay, nothing special.",
    "What a great experience!"
]

# Tokenize the input texts
# `truncation=True` ensures that texts longer than the model's max input length are truncated.
# `padding=True` pads shorter texts to the same length as the longest text in the batch.
inputs = tokenizer(texts, return_tensors="tf", truncation=True, padding=True)

# Perform inference
outputs = model(inputs)

# Get the predicted probabilities (logits are raw scores, apply softmax to get probabilities)
predictions = tf.nn.softmax(outputs.logits, axis=-1)

# Get the predicted labels
# The labels for this specific model are typically 0 for negative and 1 for positive.
predicted_labels = tf.argmax(predictions, axis=-1).numpy()

# Map numerical labels to human-readable labels
label_map = {0: "Negative", 1: "Positive"}

print("Sentiment Analysis Results:")
for i, text in enumerate(texts):
    sentiment = label_map[predicted_labels[i]]
    confidence = predictions[i][predicted_labels[i]].numpy() * 100
    print(f"- Text: '{text}'\n  Predicted Sentiment: {sentiment} ({confidence:.2f}% confidence)\n")
```

This example demonstrates a basic use case of transformers for text classification. The Hugging Face library provides a wide array of pre-trained models for various tasks, and you can also fine-tune these models on your own custom datasets to achieve even better performance for your specific needs.
