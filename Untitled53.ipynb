## Exploratory Data Analysis (EDA)

EDA helps us understand the data better before applying any machine learning models. Here are some common steps involved in EDA:

1.  **Data Loading**: Load the dataset into a suitable data structure (e.g., a Pandas DataFrame).
2.  **Data Inspection**: Check the dimensions, data types, and presence of missing values. Understand the column names and what each column represents.
3.  **Descriptive Statistics**: Calculate summary statistics (mean, median, mode, standard deviation, quartiles, etc.) for numerical features. For categorical features, look at value counts.
4.  **Data Visualization**: Create various plots to understand distributions, relationships between variables, and identify outliers:
    *   **Histograms/KDE plots**: To visualize the distribution of numerical features.
    *   **Bar plots**: To visualize the distribution of categorical features.
    *   **Box plots**: To identify outliers and understand the spread of numerical data, often grouped by categorical variables.
    *   **Scatter plots**: To visualize relationships between two numerical variables.
    *   **Correlation matrices/Heatmaps**: To understand the correlation between numerical features.
    *   **Time series plots**: If the data has a time component.
5.  **Handling Missing Values**: Decide on a strategy for missing data (e.g., imputation, deletion).
6.  **Outlier Detection**: Identify and understand the nature of outliers.
7.  **Feature Engineering (initial)**: Create new features that might be useful for modeling.

### Generating Summary Statistics

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame
data = {
    'numerical_col_1': np.random.rand(100) * 100,
    'numerical_col_2': np.random.randint(1, 50, 100),
    'categorical_col': np.random.choice(['A', 'B', 'C', 'D'], 100),
    'boolean_col': np.random.choice([True, False], 100)
}
df = pd.DataFrame(data)

print("Sample DataFrame head:")
display(df.head())
```

```python
# Generate summary statistics for numerical columns
print("\nSummary statistics for numerical columns:")
display(df.describe())
```

```python
# Generate summary statistics for all columns, including categorical and boolean
print("\nSummary statistics for all columns (including categorical and boolean):")
display(df.describe(include='all'))
```

The `df.describe()` method provides a quick overview of numerical features, showing count, mean, standard deviation, min, max, and quartile values. When `include='all'` is used, it also includes statistics for categorical and boolean columns, such as unique values, top occurring value, and its frequency.
