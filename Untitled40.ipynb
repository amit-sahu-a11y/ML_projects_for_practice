## Advanced Topic: Python Decorators

Python decorators are a powerful and advanced feature that allows you to modify or enhance functions or methods. They are essentially functions that take another function as an argument, add some functionality, and return a new function.

````python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Something is happening before the function is called.")
        result = func(*args, **kwargs)
        print("Something is happening after the function is called.")
        return result
    return wrapper

@my_decorator
def say_hello(name):
    print(f"Hello, {name}!")
    return f"Greeting for {name}"

@my_decorator
def calculate_sum(a, b):
    print(f"Calculating sum of {a} and {b}")
    return a + b

print("\n--- Calling say_hello ---")
say_hello("Alice")

print("\n--- Calling calculate_sum ---")
sum_result = calculate_sum(10, 20)
print(f"Sum result: {sum_result}")
````

### Explanation:

1.  **`my_decorator(func)`**: This is the decorator function. It takes a function `func` as an argument.
2.  **`wrapper(*args, **kwargs)`**: This is the inner function that `my_decorator` returns. It's designed to accept any arguments that the original `func` might take. Inside `wrapper`, you can add code to execute before and after `func` is called.
3.  **`@my_decorator`**: This is the syntactic sugar for applying the decorator. It's equivalent to `say_hello = my_decorator(say_hello)` and `calculate_sum = my_decorator(calculate_sum)`.

When `say_hello('Alice')` or `calculate_sum(10, 20)` is called, the `wrapper` function is actually executed, which then calls the original `say_hello` or `calculate_sum` function along with the added "before" and "after" logic.

## Combining Multiple Decorators

You can apply multiple decorators to a single function. When you do this, the decorators are applied in the order they are listed, from bottom to top (or closest to the function definition first, then moving upwards).

Consider this syntax:

```python
@decorator1
@decorator2
def my_function():
    pass
```

This is equivalent to:

```python
my_function = decorator1(decorator2(my_function))
```

Meaning `decorator2` is applied first to `my_function`, and then `decorator1` is applied to the result of `decorator2(my_function)`.

````python
class CallCountDecorator:
    def __init__(self, func):
        self.func = func
        self.call_count = 0

    def __call__(self, *args, **kwargs):
        self.call_count += 1
        print(f"\n(CallCount) Call count for '{self.func.__name__}': {self.call_count}")
        return self.func(*args, **kwargs)


def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("(MyDecorator) Something is happening before the function is called.")
        result = func(*args, **kwargs)
        print("(MyDecorator) Something is happening after the function is called.")
        return result
    return wrapper

def repeat(num_times, message="Executing function..."):
    def decorator_repeat(func):
        def wrapper(*args, **kwargs):
            print(f"(RepeatDecorator) Custom message: {message}")
            for _ in range(num_times):
                func(*args, **kwargs)
        return wrapper
    return decorator_repeat

@CallCountDecorator
@repeat(num_times=2, message="Repeating the decorated function!")
@my_decorator
def complex_task(item):
    print(f"  Executing complex task for: {item}")

print("\n--- Calling complex_task with multiple decorators ---")
complex_task("Report")
complex_task("Analysis")
````

### Explanation of Multiple Decorator Application:

In the `complex_task` example:

```python
@CallCountDecorator
@repeat(num_times=2, message="Repeating the decorated function!")
@my_decorator
def complex_task(item):
    print(f"  Executing complex task for: {item}")
```

1.  **`@my_decorator`** is applied first to the `complex_task` function.
2.  The result of `my_decorator(complex_task)` is then passed to the `repeat` decorator factory, which returns a decorator. This decorator then wraps the result from `my_decorator`.
3.  Finally, the `CallCountDecorator` class is applied to the result of `repeat` decorator. This means an instance of `CallCountDecorator` is created, wrapping the function returned by `repeat`.

When `complex_task()` is called:

*   The `__call__` method of the `CallCountDecorator` instance is invoked first.
*   Inside `CallCountDecorator.__call__`, it calls the next wrapped function (which is the `wrapper` from `repeat`).
*   Inside `repeat`'s `wrapper`, it prints its custom message and then loops `num_times`.
*   In each iteration of the `repeat` loop, it calls the innermost wrapped function (which is the `wrapper` from `my_decorator`).
*   Inside `my_decorator`'s `wrapper`, it prints its "before" message, calls the original `complex_task`, and then prints its "after" message.

This nesting explains the order of output you'll observe.

## Passing Arguments to Decorators

To pass arguments to a decorator, you need to create a decorator factory. This is a function that takes the decorator arguments and returns the actual decorator. The actual decorator then takes the function to be decorated as its argument.

````python
def repeat(num_times, message="Executing function..."):
    def decorator_repeat(func):
        def wrapper(*args, **kwargs):
            print(f"\nCustom message: {message}")
            for _ in range(num_times):
                func(*args, **kwargs)
        return wrapper
    return decorator_repeat

@repeat(num_times=3, message="Time to greet!")
def greet(name):
    print(f"Hello, {name}!")

print("\n--- Calling greet with decorator arguments and custom message ---")
greet("Bob")

@repeat(num_times=2)
def say_goodbye(name):
    print(f"Goodbye, {name}!")

print("\n--- Calling say_goodbye with default message ---")
say_goodbye("Alice")
````

### Explanation of the `repeat` decorator:

1.  **`repeat(num_times, message="Executing function...")`**: This is the decorator factory. It now takes `num_times` and an optional `message` argument.
2.  **`decorator_repeat(func)`**: This is the actual decorator that gets returned by `repeat`. It takes the function to be decorated (`greet` or `say_goodbye` in this case) as an argument.
3.  **`wrapper(*args, **kwargs)`**: This is the inner function. It now prints the `message` before calling the decorated function multiple times.
4.  **`@repeat(num_times=3, message="Time to greet!")`**: When you use this syntax, Python first calls `repeat(num_times=3, message="Time to greet!")`, which returns `decorator_repeat`. Then, `decorator_repeat` is called with `greet` as its argument.
5.  **`@repeat(num_times=2)`**: If no message is provided, the default message "Executing function..." will be used.
