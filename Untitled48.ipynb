## Advanced Linear Programming (LP)

Linear Programming (LP) is a mathematical method for determining a way to achieve the best outcome (such as maximum profit or lowest cost) in a mathematical model whose requirements are represented by linear relationships. LP is a special case of mathematical programming.

### Key Components:
1.  **Decision Variables**: These are the quantities we want to determine to optimize the objective function. They are typically denoted as $x_1, x_2, \dots, x_n$.
2.  **Objective Function**: This is a linear function that we want to maximize or minimize. It takes the form: $Z = c_1x_1 + c_2x_2 + \dots + c_nx_n$.
3.  **Constraints**: These are linear inequalities or equalities that restrict the possible values of the decision variables. They often represent resource limitations or other requirements. For example: $a_{i1}x_1 + a_{i2}x_2 + \dots + a_{in}x_n \le b_i$.
4.  **Non-negativity Constraints**: Decision variables are typically assumed to be non-negative: $x_j \ge 0$ for all $j$.

### General Form:

**Minimize** $Z = c^T x$

**Subject to:**
$Ax \le b$
$A_{eq}x = b_{eq}$
$l \le x \le u$

Where:
*   $c$ is the coefficients of the objective function.
*   $x$ is the vector of decision variables.
*   $A$ and $A_{eq}$ are matrices of coefficients for inequality and equality constraints, respectively.
*   $b$ and $b_{eq}$ are vectors of right-hand side values for inequality and equality constraints.
*   $l$ and $u$ are lower and upper bounds on the decision variables.

### Example Problem: Resource Allocation

Let's consider a simple manufacturing problem. A company produces two products, Product A and Product B. Each product requires processing time on two different machines, Machine 1 and Machine 2.

*   **Product A**: Requires 1 hour on Machine 1 and 3 hours on Machine 2.
*   **Product B**: Requires 2 hours on Machine 1 and 1 hour on Machine 2.

Available machine hours:
*   Machine 1: 10 hours per week.
*   Machine 2: 15 hours per week.

Profit per unit:
*   Product A: $6
*   Product B: $8

**Objective**: Maximize the total profit.

**Decision Variables**:
*   $x_A$: Number of units of Product A produced.
*   $x_B$: Number of units of Product B produced.

**Objective Function (Maximize Profit)**:
Maximize $Z = 6x_A + 8x_B$

**Constraints**:
*   Machine 1: $1x_A + 2x_B \le 10$
*   Machine 2: $3x_A + 1x_B \le 15$
*   Non-negativity: $x_A \ge 0, x_B \ge 0$

```python
import numpy as np
from scipy.optimize import linprog

# The objective function coefficients (c). Since linprog minimizes, we negate for maximization.
c = [-6, -8]  # Maximize 6x_A + 8x_B is equivalent to Minimize -6x_A - 8x_B

# Coefficients for the inequality constraints (A_ub)
A_ub = [
    [1, 2],  # Machine 1 constraint: 1x_A + 2x_B <= 10
    [3, 1]   # Machine 2 constraint: 3x_A + 1x_B <= 15
]

# Right-hand side values for the inequality constraints (b_ub)
b_ub = [10, 15]

# Bounds for the decision variables (x_A, x_B). Non-negative.
x_bounds = (0, None)  # x_A >= 0
y_bounds = (0, None)  # x_B >= 0
```

```python
# Solve the linear programming problem
res = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=[x_bounds, y_bounds], method='highs')

# Display the results
print("Optimization Result:")
print(res)
```

### Interpretation of Results:

*   **`fun`**: This is the optimal value of the objective function. Remember we negated the coefficients for maximization, so we need to negate this value to get the maximum profit.
*   **`x`**: This array contains the optimal values of the decision variables ($x_A, x_B$) that achieve the optimal `fun`.
*   **`message`**: A string indicating the status of the optimization. 'Optimization terminated successfully.' means a solution was found.
*   **`status`**: An integer indicating the exit status of the optimization (0 usually means success).



```python
if res.success:
    optimal_xA = res.x[0]
    optimal_xB = res.x[1]
    max_profit = -res.fun # Negate to get actual max profit

    print(f"\nOptimal production quantity of Product A: {optimal_xA:.2f} units")
    print(f"Optimal production quantity of Product B: {optimal_xB:.2f} units")
    print(f"Maximum profit: ${max_profit:.2f}")
else:
    print(f"The optimization did not converge successfully. Status: {res.status}, Message: {res.message}")
```

## Sensitivity Analysis for Linear Programming

Sensitivity analysis (or post-optimality analysis) examines how changes in the input parameters of a linear programming problem affect the optimal solution. The primary parameters we are interested in are:

1.  **Objective Function Coefficients**: How much can the profit (or cost) of a product change before the optimal production mix changes?
2.  **Right-Hand Side (RHS) of Constraints**: How much can the available resources (e.g., machine hours) change before the shadow price (value of an extra unit of resource) changes, or before the optimal solution becomes infeasible?

### Key Concepts:

*   **Shadow Price (Dual Price / Dual Value)**: The shadow price of a constraint represents the change in the optimal objective function value for a one-unit increase in the right-hand side of that constraint, assuming all other parameters remain constant. In simple terms, it's the maximum price you would be willing to pay for one additional unit of a scarce resource.
    *   In `scipy.optimize.linprog`, shadow prices for inequality constraints (`A_ub`) are given by `res.ineqlin.marginals`.

*   **Reduced Cost**: The reduced cost for a non-basic variable (a variable that is 0 in the optimal solution) indicates how much the objective function coefficient of that variable would have to improve (decrease for minimization, increase for maximization) before it would become optimal to produce that variable. For basic variables (variables that are positive in the optimal solution), the reduced cost is typically 0.
    *   In `scipy.optimize.linprog`, reduced costs are given by `res.lower.marginals` for the lower bounds and `res.upper.marginals` for the upper bounds. For variables that are positive, their 'reduced cost' (or marginal value associated with their non-negativity constraint) should be 0, meaning they are already being produced optimally.

Let's extract and interpret these values from our `res` object.

```python
# Extract shadow prices (dual values) for inequality constraints
# The order corresponds to the order of constraints in A_ub
shadow_prices = res.ineqlin.marginals

print(f"Shadow Price for Machine 1 (Constraint 1: 1x_A + 2x_B <= 10): {-shadow_prices[0]:.2f}") # Negate for maximization interpretation
print(f"Shadow Price for Machine 2 (Constraint 2: 3x_A + 1x_B <= 15): {-shadow_prices[1]:.2f}") # Negate for maximization interpretation

print("\nInterpretation:")
print(f"  - For every additional hour on Machine 1, the maximum profit could increase by ${-shadow_prices[0]:.2f}.")
print(f"  - For every additional hour on Machine 2, the maximum profit could increase by ${-shadow_prices[1]:.2f}.")
print("  (Note: These interpretations hold only within a certain range of change for the resource availability.)")
```

```python
# Extract reduced costs for decision variables
# For variables with positive values in the optimal solution (like our xA and xB here), reduced cost is 0.
# If a variable were 0, its reduced cost would indicate how much its objective coefficient needs to improve to become active.

# linprog output for 'marginals' here refers to the marginal costs associated with the bounds.
# For a variable x_j, if x_j > 0, its lower bound marginal (reduced cost) is 0.
# If x_j = 0 and its lower bound marginal is negative (for minimization), it means increasing x_j would decrease the objective.

# Let's consider the general interpretation if a variable was not produced.
# In our case, optimal_xA and optimal_xB are both positive, so their reduced costs (for the non-negativity constraint) are 0.

reduced_cost_xA = res.lower.marginals[0] if res.x[0] == 0 else 0
reduced_cost_xB = res.lower.marginals[1] if res.x[1] == 0 else 0

print(f"\nReduced Cost for Product A: {reduced_cost_xA:.2f}")
print(f"Reduced Cost for Product B: {reduced_cost_xB:.2f}")

print("\nInterpretation:")
if reduced_cost_xA == 0:
    print(f"  - Product A is being produced (4.00 units), so its reduced cost is 0.")
else:
    print(f"  - If Product A were not produced, its profit would need to increase by {reduced_cost_xA:.2f} per unit to become profitable.")

if reduced_cost_xB == 0:
    print(f"  - Product B is being produced (3.00 units), so its reduced cost is 0.")
else:
    print(f"  - If Product B were not produced, its profit would need to increase by {reduced_cost_xB:.2f} per unit to become profitable.")
```
