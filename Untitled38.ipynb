### Metropolis-Hastings Algorithm

The Metropolis-Hastings (MH) algorithm is a powerful Markov Chain Monte Carlo (MCMC) method used to draw samples from probability distributions that are difficult to sample from directly. It's particularly useful when you can calculate the unnormalized probability density function (PDF) of your target distribution, but not its normalization constant.

**How it works (in simple terms):**

1.  **Start Somewhere:** Begin with an initial state (a sample point) in the parameter space.
2.  **Propose a Move:** From the current state, propose a new state (candidate sample) using a *proposal distribution*. This proposal distribution is usually simple (e.g., a normal distribution centered at the current state).
3.  **Decide to Accept or Reject:** Calculate an *acceptance ratio*. This ratio compares the probability density of the target distribution at the proposed state to the current state, adjusted by the proposal distribution. If the proposed state is 'better' (has higher probability density), it's often accepted. If it's 'worse', it might still be accepted with a certain probability.
    *   The acceptance probability is given by:
        $$ \alpha = \min\left(1, \frac{P(x_{\text{new}}) \cdot Q(x_{\text{current}} | x_{\text{new}})}{P(x_{\text{current}}) \cdot Q(x_{\text{new}} | x_{\text{current}})}\right) $$
        Where:
        *   $P(x)$ is the target distribution's unnormalized PDF.
        *   $Q(x_{\text{new}} | x_{\text{current}})$ is the probability of proposing $x_{\text{new}}$ given $x_{\text{current}}$.

4.  **Update State:** Generate a random number between 0 and 1. If this number is less than or equal to the acceptance ratio, accept the proposed state and move there. Otherwise, reject the proposal and remain at the current state.
5.  **Repeat:** Repeat steps 2-4 many times. As the chain progresses, the samples drawn will approximate the target distribution.

**Key Concepts:**

*   **Target Distribution:** The distribution you want to sample from.
*   **Proposal Distribution:** A simpler distribution used to suggest new candidate samples.
*   **Markov Chain:** The sequence of accepted samples forms a Markov chain, where each state depends only on the previous one.
*   **Burn-in:** The initial samples of the chain are often discarded because they are still converging to the target distribution.
*   **Autocorrelation:** Samples from MCMC can be correlated. You might 'thin' the chain (take every Nth sample) to reduce this.

**When to use it:**

*   When the target distribution is complex and cannot be sampled directly.
*   When the normalization constant of the target distribution is unknown or difficult to compute.
*   Common in Bayesian statistics for posterior sampling.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Define the target distribution (e.g., a Gaussian distribution)
# We can use an unnormalized version, as MH algorithm doesn't require normalization constant.
def target_distribution(x):
    # This is an unnormalized Gaussian centered at 0 with standard deviation 2
    return np.exp(-0.5 * (x / 2)**2)

# 2. Define the proposal distribution
# A simple symmetric proposal: a normal distribution centered at the current_state
def proposal_distribution(current_state, scale=1.0):
    return np.random.normal(loc=current_state, scale=scale)

# 3. Implement the Metropolis-Hastings algorithm
def run_metropolis_hastings(target_dist, proposal_dist, initial_state, num_samples, proposal_scale=1.0):
    samples = [initial_state]
    current_state = initial_state
    accepted_count = 0

    for _ in range(num_samples - 1):
        # Propose a new state
        proposed_state = proposal_dist(current_state, proposal_scale)

        # Calculate the acceptance ratio
        # Since the proposal is symmetric (Normal centered at current_state),
        # Q(x_current | x_new) / Q(x_new | x_current) = 1
        # So the ratio simplifies to target_dist(proposed) / target_dist(current)
        acceptance_ratio = min(1, target_dist(proposed_state) / target_dist(current_state))

        # Accept or reject the proposed state
        if np.random.rand() < acceptance_ratio:
            current_state = proposed_state
            accepted_count += 1
        samples.append(current_state)

    acceptance_rate = accepted_count / (num_samples - 1) if num_samples > 1 else 0
    return np.array(samples), acceptance_rate

# --- Run the MH algorithm ---
initial_state = 0.0
num_samples = 50000
proposal_scale = 1.0 # Standard deviation of the proposal normal distribution

mh_samples, acceptance_rate = run_metropolis_hastings(target_distribution, proposal_distribution, initial_state, num_samples, proposal_scale)

# --- Visualization ---
plt.figure(figsize=(10, 6))
sns.histplot(mh_samples[1000:], bins=50, stat='density', color='skyblue', label='MH Samples (after burn-in)')

# Plot the true target distribution for comparison
x = np.linspace(-10, 10, 500)
plt.plot(x, target_distribution(x) / np.trapezoid(target_distribution(x), x), color='red', linestyle='--', label='True Target Distribution (Normalized)')

plt.title('Metropolis-Hastings Sampling of a Gaussian Distribution')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

print(f"Mean of MH samples: {np.mean(mh_samples[1000:]):.3f}")
print(f"Standard deviation of MH samples: {np.std(mh_samples[1000:]):.3f}")
print(f"Acceptance rate: {acceptance_rate:.3f}")
```
