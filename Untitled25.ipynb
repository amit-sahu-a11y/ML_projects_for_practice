## 1. Data Model Definition: Property and Client Classes

We'll define Python classes to represent the fundamental entities in our real estate CRM: `Property` and `Client`. Each class will have attributes relevant to its role.

```python
class Property:
    def __init__(self, property_id, address, city, state, zip_code, price, bedrooms, bathrooms, area_sqft, status='Available'):
        self.property_id = property_id
        self.address = address
        self.city = city
        self.state = state
        self.zip_code = zip_code
        self.price = price
        self.bedrooms = bedrooms
        self.bathrooms = bathrooms
        self.area_sqft = area_sqft
        self.status = status  # e.g., 'Available', 'Under Contract', 'Sold'

    def __repr__(self):
        return f"Property(ID: {self.property_id}, Address: {self.address}, Price: ${self.price:,.2f}, Status: {self.status})"

class Client:
    def __init__(self, client_id, first_name, last_name, email, phone, budget_min=None, budget_max=None, preferred_bedrooms=None, preferred_bathrooms=None):
        self.client_id = client_id
        self.first_name = first_name
        self.last_name = last_name
        self.email = email
        self.phone = phone
        self.budget_min = budget_min
        self.budget_max = budget_max
        self.preferred_bedrooms = preferred_bedrooms
        self.preferred_bathrooms = preferred_bathrooms
        self.interested_properties = [] # List of property_ids

    def __repr__(self):
        return f"Client(ID: {self.client_id}, Name: {self.first_name} {self.last_name}, Email: {self.email})"

    def add_interested_property(self, property_id):
        if property_id not in self.interested_properties:
            self.interested_properties.append(property_id)
            print(f"Property {property_id} added to {self.first_name}'s interested list.")
        else:
            print(f"Client {self.first_name} is already interested in property {property_id}.")

    def remove_interested_property(self, property_id):
        if property_id in self.interested_properties:
            self.interested_properties.remove(property_id)
            print(f"Property {property_id} removed from {self.first_name}'s interested list.")
        else:
            print(f"Client {self.first_name} is not interested in property {property_id}.")
```

## 2. Data Storage and Basic Operations: CRMManager Class

To effectively manage our `Property` and `Client` objects, we'll create a `CRMManager` class. This class will encapsulate the logic for storing, adding, retrieving, updating, and deleting these entities, providing a centralized way to interact with our CRM data.

```python
class CRMManager:
    def __init__(self):
        self.properties = []
        self.clients = []
        self._next_property_id = 1
        self._next_client_id = 1

    def _generate_property_id(self):
        prop_id = f"P{self._next_property_id:03d}"
        self._next_property_id += 1
        return prop_id

    def _generate_client_id(self):
        client_id = f"C{self._next_client_id:03d}"
        self._next_client_id += 1
        return client_id

    def add_property(self, address, city, state, zip_code, price, bedrooms, bathrooms, area_sqft, status='Available'):
        property_id = self._generate_property_id()
        new_property = Property(property_id, address, city, state, zip_code, price, bedrooms, bathrooms, area_sqft, status)
        self.properties.append(new_property)
        print(f"Added property: {new_property}")
        return new_property

    def get_property(self, property_id):
        for prop in self.properties:
            if prop.property_id == property_id:
                return prop
        return None

    def list_properties(self):
        if not self.properties:
            print("No properties in the CRM.")
            return []
        print("--- All Properties ---")
        for prop in self.properties:
            print(prop)
        return self.properties

    def update_property_status(self, property_id, new_status):
        prop = self.get_property(property_id)
        if prop:
            prop.status = new_status
            print(f"Property {property_id} status updated to {new_status}")
            return True
        print(f"Property with ID {property_id} not found.")
        return False

    def add_client(self, first_name, last_name, email, phone, budget_min=None, budget_max=None, preferred_bedrooms=None, preferred_bathrooms=None):
        client_id = self._generate_client_id()
        new_client = Client(client_id, first_name, last_name, email, phone, budget_min, budget_max, preferred_bedrooms, preferred_bathrooms)
        self.clients.append(new_client)
        print(f"Added client: {new_client}")
        return new_client

    def get_client(self, client_id):
        for client in self.clients:
            if client.client_id == client_id:
                return client
        return None

    def list_clients(self):
        if not self.clients:
            print("No clients in the CRM.")
            return []
        print("--- All Clients ---")
        for client in self.clients:
            print(client)
        return self.clients

    def client_express_interest(self, client_id, property_id):
        client = self.get_client(client_id)
        prop = self.get_property(property_id)
        if client and prop:
            client.add_interested_property(property_id)
            return True
        print("Client or Property not found.")
        return False

    def client_remove_interest(self, client_id, property_id):
        client = self.get_client(client_id)
        if client:
            client.remove_interested_property(property_id)
            return True
        print("Client not found.")
        return False
```

## Demo: Using the CRMManager

Now let's use our `CRMManager` to add properties and clients, list them, and demonstrate some basic operations.

```python
# Initialize the CRM Manager
manager = CRMManager()

# Add some properties using the manager
prop1 = manager.add_property('123 Main St', 'Anytown', 'CA', '90210', 500000, 3, 2, 1800)
prop2 = manager.add_property('456 Oak Ave', 'Otherville', 'NY', '10001', 750000, 4, 3, 2500)
prop3 = manager.add_property('789 Pine Ln', 'Anytown', 'CA', '90210', 300000, 2, 1, 1200)

# Add some clients using the manager
cli1 = manager.add_client('Alice', 'Smith', 'alice@example.com', '555-123-4567', 400000, 600000, 3, 2)
cli2 = manager.add_client('Bob', 'Johnson', 'bob@example.com', '555-987-6543', 600000, 800000, 4, 3)

print("\n")

# List all properties and clients
manager.list_properties()
manager.list_clients()

print("\n")

# Demonstrate updating a property status
manager.update_property_status(prop1.property_id, 'Under Contract')
print(manager.get_property(prop1.property_id))

# Demonstrate client interest
print("\n--- Client Interest Management ---")
manager.client_express_interest(cli1.client_id, prop1.property_id)
manager.client_express_interest(cli1.client_id, prop3.property_id)
manager.client_express_interest(cli2.client_id, prop2.property_id)

print(f"\nClient {cli1.first_name} interested in: {manager.get_client(cli1.client_id).interested_properties}")
print(f"Client {cli2.first_name} interested in: {manager.get_client(cli2.client_id).interested_properties}")

manager.client_remove_interest(cli1.client_id, prop1.property_id)
print(f"Client {cli1.first_name} interested after removal: {manager.get_client(cli1.client_id).interested_properties}")
```

## 3. Advanced Search and Filtering with CRMManager

To make our CRM truly useful, we need robust search and filtering capabilities. We'll extend the `CRMManager` to include methods for searching properties by various attributes and for converting our data into Pandas DataFrames for easier querying and analysis.

```python
class CRMManager(CRMManager):
    # Inherit from the previous CRMManager and extend it
    def search_properties(self, min_price=None, max_price=None, min_bedrooms=None, min_bathrooms=None, status=None, city=None):
        results = self.properties
        if min_price is not None: results = [p for p in results if p.price >= min_price]
        if max_price is not None: results = [p for p in results if p.price <= max_price]
        if min_bedrooms is not None: results = [p for p in results if p.bedrooms >= min_bedrooms]
        if min_bathrooms is not None: results = [p for p in results if p.bathrooms >= min_bathrooms]
        if status is not None: results = [p for p in results if p.status == status]
        if city is not None: results = [p for p in results if p.city.lower() == city.lower()]

        if not results:
            print("No properties found matching the criteria.")
        else:
            print("--- Search Results (Properties) ---")
            for prop in results:
                print(prop)
        return results

    def search_clients(self, budget_min=None, budget_max=None, preferred_bedrooms=None, preferred_bathrooms=None, first_name=None, last_name=None):
        results = self.clients
        if budget_min is not None: results = [c for c in results if c.budget_max is not None and c.budget_max >= budget_min]
        if budget_max is not None: results = [c for c in results if c.budget_min is not None and c.budget_min <= budget_max]
        if preferred_bedrooms is not None: results = [c for c in results if c.preferred_bedrooms is not None and c.preferred_bedrooms >= preferred_bedrooms]
        if preferred_bathrooms is not None: results = [c for c in results if c.preferred_bathrooms is not None and c.preferred_bathrooms >= preferred_bathrooms]
        if first_name is not None: results = [c for c in results if c.first_name.lower() == first_name.lower()]
        if last_name is not None: results = [c for c in results if c.last_name.lower() == last_name.lower()]

        if not results:
            print("No clients found matching the criteria.")
        else:
            print("--- Search Results (Clients) ---")
            for client in results:
                print(client)
        return results

    def properties_to_dataframe(self):
        property_data = []
        for prop in self.properties:
            property_data.append({
                'property_id': prop.property_id,
                'address': prop.address,
                'city': prop.city,
                'state': prop.state,
                'zip_code': prop.zip_code,
                'price': prop.price,
                'bedrooms': prop.bedrooms,
                'bathrooms': prop.bathrooms,
                'area_sqft': prop.area_sqft,
                'status': prop.status
            })
        return pd.DataFrame(property_data)

    def clients_to_dataframe(self):
        client_data = []
        for client in self.clients:
            client_data.append({
                'client_id': client.client_id,
                'first_name': client.first_name,
                'last_name': client.last_name,
                'email': client.email,
                'phone': client.phone,
                'budget_min': client.budget_min,
                'budget_max': client.budget_max,
                'preferred_bedrooms': client.preferred_bedrooms,
                'preferred_bathrooms': client.preferred_bathrooms,
                'interested_properties': ', '.join(client.interested_properties) # Store as comma-separated string
            })
        return pd.DataFrame(client_data)
```
