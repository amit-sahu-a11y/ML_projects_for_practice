## Federated Learning: Training Models While Preserving Privacy

Federated Learning is a distributed machine learning approach that enables training an algorithm across multiple decentralized edge devices or servers holding local data samples, without exchanging them. This contrasts with traditional centralized machine learning techniques where all datasets are uploaded to one server.

### Key Concepts:
1.  **Decentralized Data**: Data remains on local devices (clients) and is not shared with a central server.
2.  **Model Aggregation**: Clients train local models using their own data, and only model updates (e.g., gradients or weights) are sent to a central server.
3.  **Privacy Preservation**: Raw data never leaves the client's device, significantly enhancing data privacy and security.
4.  **Communication Efficiency**: Only small model updates are transmitted, reducing communication overhead compared to sending entire datasets.

### Use Cases:
*   **Healthcare**: Training models on patient data from various hospitals without violating privacy regulations.
*   **Mobile Devices**: Improving predictive text or voice recognition models directly on user devices.
*   **Financial Services**: Detecting fraud by training models on sensitive transaction data from different banks.

Let's look at a simplified conceptual example using PyTorch to illustrate the core idea of federated averaging.

```python
# First, let's install the necessary libraries.
# We'll use PyTorch for the model and demonstrate the concept.
# For a full-fledged federated learning framework, you might use libraries like PySyft or Flower.
!pip install torch torchvision

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import numpy as np

print('PyTorch and related libraries installed and imported successfully.')
```

### 1. Define a Simple Neural Network Model

```python
class SimpleModel(nn.Module):
    def __init__(self):
        super(SimpleModel, self).__init__()
        self.fc1 = nn.Linear(10, 5)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(5, 1)

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        return x

# Initialize the global model
global_model = SimpleModel()
print('Global model initialized:')
print(global_model)
```

### 2. Simulate Data on Multiple Clients and Federated Averaging

In this simplified example, we'll create some dummy data for each client and then simulate the federated averaging process. Each client trains a local model, sends its updated weights to a central server, and the server aggregates these updates to create a new global model.

```python
# Simulation parameters
num_clients = 3
epochs_per_client = 1
learning_rate = 0.01

# Create dummy data for each client
client_data = []
for i in range(num_clients):
    # Each client has its own unique data distribution
    X = torch.randn(100, 10) + i * 2  # Features
    y = (X[:, 0] * 0.5 + X[:, 1] * 0.2 + torch.randn(100) * 0.1).unsqueeze(1) # Labels
    client_data.append(TensorDataset(X, y))

# List to store average loss per federated round for plotting
federated_round_losses = []

# Federated Learning Round
print(f"\nStarting Federated Learning with {num_clients} clients...")

for federated_round in range(5): # Simulate 5 rounds
    print(f"\n--- Federated Round {federated_round + 1} ---")
    client_models_weights = []
    current_round_client_losses = []

    # 1. Clients download global model and train locally
    for i, dataset in enumerate(client_data):
        print(f"Client {i+1} training...")
        local_model = SimpleModel() # Each client gets a fresh instance
        local_model.load_state_dict(global_model.state_dict()) # Initialize with global model weights

        dataloader = DataLoader(dataset, batch_size=10, shuffle=True)
        optimizer = optim.SGD(local_model.parameters(), lr=learning_rate)
        criterion = nn.MSELoss()

        for epoch in range(epochs_per_client):
            for X_batch, y_batch in dataloader:
                optimizer.zero_grad()
                outputs = local_model(X_batch)
                loss = criterion(outputs, y_batch)
                loss.backward()
                optimizer.step()

        current_round_client_losses.append(loss.item())
        print(f"Client {i+1} finished. Local loss: {loss.item():.4f}")
        client_models_weights.append(local_model.state_dict()) # Store client's updated weights

    # Calculate average loss for the current round
    avg_round_loss = np.mean(current_round_client_losses)
    federated_round_losses.append(avg_round_loss)
    print(f"Average loss for Round {federated_round + 1}: {avg_round_loss:.4f}")

    # 2. Server aggregates client updates (Federated Averaging)
    print("Server aggregating models...")
    global_state_dict = global_model.state_dict()
    for key in global_state_dict.keys():
        global_state_dict[key] = torch.stack([client_weights[key] for client_weights in client_models_weights]).mean(dim=0)
    global_model.load_state_dict(global_state_dict)
    print("Server aggregation complete. New global model updated.")

print("\nFederated Learning simulation complete!")
print("Final global model weights (sample):", global_model.fc1.weight[0, :5])
```

### 3. Visualize Federated Learning Loss Over Rounds

Let's plot the average loss recorded at the end of each federated round to see how the model converges.

```python
import matplotlib.pyplot as plt

# Ensure federated_round_losses is a list of numbers
# (it was stored as numpy floats, which is fine for plotting)

rounds = range(1, len(federated_round_losses) + 1)

plt.figure(figsize=(8, 5))
plt.plot(rounds, federated_round_losses, marker='o', linestyle='-', color='b')
plt.title('Average Federated Loss Over Rounds')

plt.xlabel('Federated Round')

plt.ylabel('Average Loss (MSE)')

plt.xticks(rounds)

plt.grid(True)

plt.show()

print("Loss visualization complete.")
```

2.  **`num_clients`**: Defines how many independent data sources participate in the training.
3.  **`client_data`**: Simulates diverse local datasets. In a real-world scenario, this data would reside on different physical devices.
4.  **Federated Round Loop**: Represents an iteration of the federated learning process.
    *   **Client Training**: Each client (`local_model`) gets a copy of the current `global_model`, trains on its own `dataset` for `epochs_per_client`, and then sends its updated model weights back.
    *   **Server Aggregation**: The central server collects all updated client model weights (`client_models_weights`). It then performs **Federated Averaging** by calculating the mean of all corresponding weight tensors from the clients to form the new `global_model` weights.

This example demonstrates how model parameters can be collaboratively learned without any raw data ever leaving the client devices, which is the core principle of Federated Learning.

### 6. Implementing FedProx for Non-IID Data

**FedProx** is a popular variant of Federated Averaging (FedAvg) specifically designed to address challenges posed by **Non-Independent and Identically Distributed (Non-IID)** data across clients. In FedAvg, clients train locally without any regularization, which can lead to significant *client drift* (local models diverging too much from the global model) when local datasets are heterogeneous.

FedProx introduces a **proximal term** to each client's local objective function. This term penalizes the deviation of the local model's parameters from the global model's parameters received at the beginning of the round. The modified local objective for client `k` at round `t` becomes:

$$ \min_{w_k} \mathcal{L}_k(w_k) + \frac{\mu}{2} \|w_k - w^t\|^2 $$

Where:
*   $\mathcal{L}_k(w_k)$ is the standard local loss function (e.g., MSE) for client `k`.
*   $w_k$ are the local model parameters of client `k`.
*   $w^t$ are the global model parameters received from the server at the beginning of round `t`.
*   $\mu$ is the **proximal term coefficient** (a hyperparameter), which controls the strength of the regularization. A larger $\mu$ keeps local models closer to the global model, reducing client drift but potentially slowing down adaptation to local data.

This proximal term helps to:
1.  **Reduce client drift**: By encouraging local models to stay close to the global model, it prevents them from deviating too much, especially with non-IID data.
2.  **Improve convergence**: It provides a more stable optimization path for the global model.
3.  **Enhance generalization**: By mitigating client drift, it can lead to a more robust global model that generalizes better across all clients.

```python
# To implement FedProx, we need to modify the client training loop in cell `58efe112`.
# We will add a 'mu' hyperparameter and include the proximal term in the loss calculation.

# Let's define the new simulation parameters, including the FedProx mu parameter
# You might want to copy the entire content of cell `58efe112` here and modify it,
# or modify cell `58efe112` directly. For demonstration, I'll generate a new modified cell.

# NOTE: For a clean run, you would re-execute all cells from the beginning
# after making these changes to ensure the global_model is reset.

# This cell replaces the functionality of the original simulation cell (58efe112)
# with FedProx implementation.

# Simulation parameters (updated for FedProx)
num_clients = 3
epochs_per_client = 1
learning_rate = 0.01
mu = 0.01 # FedProx proximal term coefficient - new hyperparameter

# Create dummy data for each client (re-generating for clarity if running this cell independently)
client_data = []
for i in range(num_clients):
    # Each client has its own unique data distribution
    X = torch.randn(100, 10) + i * 2  # Features
    y = (X[:, 0] * 0.5 + X[:, 1] * 0.2 + torch.randn(100) * 0.1).unsqueeze(1) # Labels
    client_data.append(TensorDataset(X, y))

# Reset global model for a fresh run with FedProx
global_model = SimpleModel() # Re-initialize the global model

# List to store average loss per federated round for plotting
federated_round_losses_fedprox = [] # Use a new list to distinguish from previous run

# Federated Learning Round with FedProx
print(f"\nStarting Federated Learning with FedProx (mu={mu}) and {num_clients} clients...")

for federated_round in range(5): # Simulate 5 rounds
    print(f"\n--- Federated Round {federated_round + 1} (FedProx) ---")
    client_models_weights = []
    current_round_client_losses = []

    # Store the initial global model parameters for proximal term calculation
    # These parameters are fixed for the current round's local training
    global_model_param_list = [p.data.clone().detach() for p in global_model.parameters()]

    # 1. Clients download global model and train locally
    for i, dataset in enumerate(client_data):
        print(f"Client {i+1} training...")
        local_model = SimpleModel() # Each client gets a fresh instance
        local_model.load_state_dict(global_model.state_dict()) # Initialize with global model weights

        dataloader = DataLoader(dataset, batch_size=10, shuffle=True)
        optimizer = optim.SGD(local_model.parameters(), lr=learning_rate)
        criterion = nn.MSELoss() # Base loss criterion

        for epoch in range(epochs_per_client):
            for X_batch, y_batch in dataloader:
                optimizer.zero_grad()
                outputs = local_model(X_batch)
                mse_loss = criterion(outputs, y_batch)

                # Calculate proximal term for FedProx
                proximal_term = 0.0
                for local_param, global_param in zip(local_model.parameters(), global_model_param_list):
                    proximal_term += torch.norm(local_param - global_param, 2) ** 2

                # Total loss = MSE loss + (mu / 2) * proximal_term
                loss = mse_loss + (mu / 2) * proximal_term
                loss.backward()
                optimizer.step()

        current_round_client_losses.append(mse_loss.item()) # Store only the MSE part for comparison
        print(f"Client {i+1} finished. Local MSE loss: {mse_loss.item():.4f}")
        client_models_weights.append(local_model.state_dict()) # Store client's updated weights

    # Calculate average MSE loss for the current round
    avg_round_loss = np.mean(current_round_client_losses)
    federated_round_losses_fedprox.append(avg_round_loss)
    print(f"Average MSE loss for Round {federated_round + 1} (FedProx): {avg_round_loss:.4f}")

    # 2. Server aggregates client updates (Federated Averaging)
    print("Server aggregating models...")
    global_state_dict = global_model.state_dict()
    for key in global_state_dict.keys():
        global_state_dict[key] = torch.stack([client_weights[key] for client_weights in client_models_weights]).mean(dim=0)
    global_model.load_state_dict(global_state_dict)
    print("Server aggregation complete. New global model updated.")

print("\nFederated Learning simulation with FedProx complete!")
print("Final global model weights (sample):", global_model.fc1.weight[0, :5])
```

Let's visualize the loss from this FedProx run and compare it with the previous FedAvg run. Remember that `federated_round_losses` stores the FedAvg losses and `federated_round_losses_fedprox` stores the FedProx losses.

```python
import matplotlib.pyplot as plt

rounds_fedavg = range(1, len(federated_round_losses) + 1)
rounds_fedprox = range(1, len(federated_round_losses_fedprox) + 1)

plt.figure(figsize=(10, 6))
plt.plot(rounds_fedavg, federated_round_losses, marker='o', linestyle='-', color='b', label='FedAvg Average MSE Loss')
plt.plot(rounds_fedprox, federated_round_losses_fedprox, marker='x', linestyle='--', color='r', label=f'FedProx (mu={mu}) Average MSE Loss')

plt.title('Comparison of Average Federated Loss Over Rounds (FedAvg vs. FedProx)')
plt.xlabel('Federated Round')
plt.ylabel('Average Loss (MSE)')
# Fix: Provide a range of integers for xticks
plt.xticks(range(1, max(rounds_fedavg[-1], rounds_fedprox[-1]) + 1))
plt.grid(True)
plt.legend()
plt.show()

print("FedProx loss visualization complete.")
```

### 5. Calculate Root Mean Squared Error (RMSE) on the Test Set

Since our model performs regression, a more appropriate metric than classification accuracy is Root Mean Squared Error (RMSE). RMSE is the square root of the MSE, making it interpretable in the same units as the predicted target variable.

```python
# Calculate RMSE from the final_test_loss (which was MSE)
rmse = np.sqrt(final_test_loss)
print(f"Root Mean Squared Error (RMSE) on test set: {rmse:.4f}")
```
