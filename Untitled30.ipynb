## Federated Learning: Training Models While Preserving Privacy

Federated Learning is a distributed machine learning approach that enables training an algorithm across multiple decentralized edge devices or servers holding local data samples, without exchanging them. This contrasts with traditional centralized machine learning techniques where all datasets are uploaded to one server.

### Key Concepts:
1.  **Decentralized Data**: Data remains on local devices (clients) and is not shared with a central server.
2.  **Model Aggregation**: Clients train local models using their own data, and only model updates (e.g., gradients or weights) are sent to a central server.
3.  **Privacy Preservation**: Raw data never leaves the client's device, significantly enhancing data privacy and security.
4.  **Communication Efficiency**: Only small model updates are transmitted, reducing communication overhead compared to sending entire datasets.

### Use Cases:
*   **Healthcare**: Training models on patient data from various hospitals without violating privacy regulations.
*   **Mobile Devices**: Improving predictive text or voice recognition models directly on user devices.
*   **Financial Services**: Detecting fraud by training models on sensitive transaction data from different banks.

Let's look at a simplified conceptual example using PyTorch to illustrate the core idea of federated averaging.

```python
# First, let's install the necessary libraries.
# We'll use PyTorch for the model and demonstrate the concept.
# For a full-fledged federated learning framework, you might use libraries like PySyft or Flower.
!pip install torch torchvision

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import numpy as np

print('PyTorch and related libraries installed and imported successfully.')
```

### 1. Define a Simple Neural Network Model

```python
class SimpleModel(nn.Module):
    def __init__(self):
        super(SimpleModel, self).__init__()
        self.fc1 = nn.Linear(10, 5)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(5, 1)

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        return x

# Initialize the global model
global_model = SimpleModel()
print('Global model initialized:')
print(global_model)
```

### 2. Simulate Data on Multiple Clients and Federated Averaging

In this simplified example, we'll create some dummy data for each client and then simulate the federated averaging process. Each client trains a local model, sends its updated weights to a central server, and the server aggregates these updates to create a new global model.

```python
# Simulation parameters
num_clients = 3
epochs_per_client = 1
learning_rate = 0.01

# Create dummy data for each client
client_data = []
for i in range(num_clients):
    # Each client has its own unique data distribution
    X = torch.randn(100, 10) + i * 2  # Features
    y = (X[:, 0] * 0.5 + X[:, 1] * 0.2 + torch.randn(100) * 0.1).unsqueeze(1) # Labels
    client_data.append(TensorDataset(X, y))

# List to store average loss per federated round for plotting
federated_round_losses = []

# Federated Learning Round
print(f"\nStarting Federated Learning with {num_clients} clients...")

for federated_round in range(5): # Simulate 5 rounds
    print(f"\n--- Federated Round {federated_round + 1} ---")
    client_models_weights = []
    current_round_client_losses = []

    # 1. Clients download global model and train locally
    for i, dataset in enumerate(client_data):
        print(f"Client {i+1} training...")
        local_model = SimpleModel() # Each client gets a fresh instance
        local_model.load_state_dict(global_model.state_dict()) # Initialize with global model weights

        dataloader = DataLoader(dataset, batch_size=10, shuffle=True)
        optimizer = optim.SGD(local_model.parameters(), lr=learning_rate)
        criterion = nn.MSELoss()

        for epoch in range(epochs_per_client):
            for X_batch, y_batch in dataloader:
                optimizer.zero_grad()
                outputs = local_model(X_batch)
                loss = criterion(outputs, y_batch)
                loss.backward()
                optimizer.step()

        current_round_client_losses.append(loss.item())
        print(f"Client {i+1} finished. Local loss: {loss.item():.4f}")
        client_models_weights.append(local_model.state_dict()) # Store client's updated weights

    # Calculate average loss for the current round
    avg_round_loss = np.mean(current_round_client_losses)
    federated_round_losses.append(avg_round_loss)
    print(f"Average loss for Round {federated_round + 1}: {avg_round_loss:.4f}")

    # 2. Server aggregates client updates (Federated Averaging)
    print("Server aggregating models...")
    global_state_dict = global_model.state_dict()
    for key in global_state_dict.keys():
        global_state_dict[key] = torch.stack([client_weights[key] for client_weights in client_models_weights]).mean(dim=0)
    global_model.load_state_dict(global_state_dict)
    print("Server aggregation complete. New global model updated.")

print("\nFederated Learning simulation complete!")
print("Final global model weights (sample):", global_model.fc1.weight[0, :5])
```

### 3. Visualize Federated Learning Loss Over Rounds

Let's plot the average loss recorded at the end of each federated round to see how the model converges.

```python
import matplotlib.pyplot as plt

# Ensure federated_round_losses is a list of numbers
# (it was stored as numpy floats, which is fine for plotting)

rounds = range(1, len(federated_round_losses) + 1)

plt.figure(figsize=(8, 5))
plt.plot(rounds, federated_round_losses, marker='o', linestyle='-', color='b')
plt.title('Average Federated Loss Over Rounds')
plt.xlabel('Federated Round')
plt.ylabel('Average Loss (MSE)')
plt.xticks(rounds)
plt.grid(True)
plt.show()

print("Loss visualization complete.")
```

### Explanation of the Code:

1.  **`SimpleModel`**: A basic neural network with one hidden layer, used as the machine learning model to be trained.
2.  **`num_clients`**: Defines how many independent data sources participate in the training.
3.  **`client_data`**: Simulates diverse local datasets. In a real-world scenario, this data would reside on different physical devices.
4.  **Federated Round Loop**: Represents an iteration of the federated learning process.
    *   **Client Training**: Each client (`local_model`) gets a copy of the current `global_model`, trains on its own `dataset` for `epochs_per_client`, and then sends its updated model weights back.
    *   **Server Aggregation**: The central server collects all updated client model weights (`client_models_weights`). It then performs **Federated Averaging** by calculating the mean of all corresponding weight tensors from the clients to form the new `global_model` weights.

This example demonstrates how model parameters can be collaboratively learned without any raw data ever leaving the client devices, which is the core principle of Federated Learning.
