## Advanced Graphs: Exploring Complex Relationships

Graphs are powerful mathematical structures used to model relationships between entities. While simple graphs might represent basic connections, **advanced graphs** delve into more complex structures, algorithms, and analyses to extract deeper insights from interconnected data. They are crucial in fields like social network analysis, bioinformatics, transportation networks, recommendation systems, and many more.

### Key Concepts in Advanced Graph Theory:

*   **Nodes (Vertices):** The entities in the graph.
*   **Edges (Links):** The connections or relationships between nodes.
*   **Directed vs. Undirected Graphs:** Edges can have a direction (e.g., A likes B, but B doesn't necessarily like A) or be undirected (e.g., A is friends with B, implies B is friends with A).
*   **Weighted Graphs:** Edges can have associated numerical values (weights), representing strength, cost, distance, etc.
*   **Cycles:** Paths in a graph that start and end at the same node.
*   **Connectivity:** How well connected the graph is (e.g., path between any two nodes).
*   **Centrality Measures:** Metrics to identify important nodes in a network (e.g., Degree, Betweenness, Closeness, Eigenvector Centrality).
*   **Community Detection:** Algorithms to find groups of nodes that are more densely connected to each other than to the rest of the network.
*   **Pathfinding Algorithms:** Algorithms like Dijkstra's or A* to find the shortest or most efficient path between two nodes.

### Introducing NetworkX

`NetworkX` is a Python library for the creation, manipulation, and study of the structure, dynamics, and functions of complex networks. It provides data structures for graphs, digraphs, and multigraphs, and a rich set of algorithms to analyze them.

```python
# First, let's install the NetworkX library if it's not already installed.
# This line is commented out as it's typically run once in a notebook setup.
# !pip install networkx matplotlib

import networkx as nx
import matplotlib.pyplot as plt

# Create an empty undirected graph
G = nx.Graph()

# Add nodes
G.add_nodes_from([1, 2, 3, 4, 5, 6])

# Add edges
G.add_edges_from([(1, 2), (1, 3), (2, 4), (3, 4), (4, 5), (5, 6), (6, 3)])

print(f"Number of nodes: {G.number_of_nodes()}")
print(f"Number of edges: {G.number_of_edges()}")
```

### Graph Properties and Basic Analysis

Once a graph is created, we can inspect its basic properties. For instance, we can find the degree of each node (the number of connections it has), check for connected components, or identify isolated nodes.

```python
# Get the degree of each node
degrees = dict(G.degree())
print(f"Node degrees: {degrees}")

# Check if the graph is connected
print(f"Is the graph connected? {nx.is_connected(G)}")

# Find the number of connected components (if not connected)
if not nx.is_connected(G):
    print(f"Number of connected components: {nx.number_connected_components(G)}")

# Example of adding an isolated node to demonstrate non-connected graph
G.add_node(7)
print(f"Is the graph connected after adding node 7? {nx.is_connected(G)}")
print(f"Number of connected components: {nx.number_connected_components(G)}")

# Remove node 7 to revert to a connected graph for further examples
G.remove_node(7)
```

### Graph Visualization

Visualizing graphs helps in understanding their structure and identifying patterns. NetworkX integrates well with Matplotlib for drawing graphs.

```python
# Draw the graph
plt.figure(figsize=(8, 6))
pos = nx.spring_layout(G)  # Choose a layout algorithm (e.g., spring_layout, circular_layout, shell_layout)
nx.draw(G, pos, with_labels=True, node_color='lightblue', node_size=1500, edge_color='gray', linewidths=1, font_size=15)
plt.title('Simple Undirected Graph')
plt.show()
```

### Centrality Measures: Identifying Important Nodes

Centrality measures help us understand the relative importance of nodes within a network. Different measures capture different aspects of 'importance'.

*   **Degree Centrality:** Measures the number of direct connections a node has. Nodes with high degree centrality are often considered 'hubs'.
*   **Betweenness Centrality:** Measures how often a node lies on the shortest path between other nodes. Nodes with high betweenness centrality act as 'bridges' or 'gatekeepers' in the network.
*   **Closeness Centrality:** Measures how close a node is to all other nodes in the network (inverse of the sum of shortest path distances). Nodes with high closeness centrality can quickly spread information.
*   **Eigenvector Centrality:** Measures a node's influence based on its connections to other highly connected nodes. A node connected to many 'important' nodes will have higher eigenvector centrality.

```python
# Calculate various centrality measures
degree_centrality = nx.degree_centrality(G)
betweenness_centrality = nx.betweenness_centrality(G)
closeness_centrality = nx.closeness_centrality(G)
eigenvector_centrality = nx.eigenvector_centrality(G, max_iter=1000) # max_iter to prevent convergence warning

print("\nCentrality Measures:")
for node in G.nodes():
    print(f"Node {node}:")
    print(f"  Degree Centrality: {degree_centrality[node]:.3f}")
    print(f"  Betweenness Centrality: {betweenness_centrality[node]:.3f}")
    print(f"  Closeness Centrality: {closeness_centrality[node]:.3f}")
    print(f"  Eigenvector Centrality: {eigenvector_centrality[node]:.3f}")

# Visualize nodes colored by their betweenness centrality
node_colors_betweenness = [betweenness_centrality[node] for node in G.nodes()]

plt.figure(figsize=(10, 8))
nodes = nx.draw_networkx_nodes(G, pos, node_color=node_colors_betweenness, cmap=plt.cm.viridis, node_size=2000)
nx.draw_networkx_edges(G, pos, edge_color='gray', alpha=0.7)
nx.draw_networkx_labels(G, pos, font_size=15, font_color='black')
plt.colorbar(nodes, orientation='horizontal', label='Betweenness Centrality')
plt.title('Graph Nodes Colored by Betweenness Centrality')
plt.show()
```

### Community Detection

Community detection is the task of identifying groups of nodes that are more densely connected to each other than to the rest of the network. These groups are often referred to as 'communities' or 'clusters'. Community detection algorithms are widely used in social network analysis, biological networks, and other complex systems to uncover underlying structures and functional modules.

One popular algorithm for community detection is the **Louvain method**. It's an iterative algorithm that optimizes the modularity of a partition of the network. Modularity is a measure of the strength of division of a network into modules.

```python
# Install the python-louvain library for community detection
# This line is commented out as it's typically run once in a notebook setup.
# !pip install python-louvain

import community as co

# Find communities using the Louvain method
# This returns a dictionary where keys are nodes and values are their community assignments
partition = co.best_partition(G)

print("Community partition:")
for node, community_id in partition.items():
    print(f"Node {node}: Community {community_id}")

# Get the modularity of the partition
modularity = co.modularity(partition, G)
print(f"\nModularity of the partition: {modularity:.3f}")
```

### Visualizing Communities

To visualize the detected communities, we can color the nodes based on their community assignment. This allows us to visually inspect how well the communities are separated within the graph.

```python
# Prepare node colors based on community assignment
# We'll map community IDs to numerical values for coloring
community_colors = [partition[node] for node in G.nodes()]

plt.figure(figsize=(10, 8))
pos = nx.spring_layout(G)  # Use the same layout for consistency

nodes = nx.draw_networkx_nodes(G, pos, node_color=community_colors, cmap=plt.cm.jet, node_size=2000)
nx.draw_networkx_edges(G, pos, edge_color='gray', alpha=0.7)
nx.draw_networkx_labels(G, pos, font_size=15, font_color='black')

plt.colorbar(nodes, orientation='horizontal', label='Community ID')
plt.title('Graph Nodes Colored by Detected Community')
plt.show()
```

### Interpreting the Results

In the visualization, nodes belonging to the same community are depicted with similar colors. The modularity score indicates how well the graph is partitioned into communities; higher values suggest a stronger community structure. By examining the communities, we can gain insights into the natural groupings or sub-structures present within the network.
